### RoBERTa Dataset 與模型

**執行類型：必跑。**

定義自訂 Dataset 與多任務 RoBERTa 架構，作為後續訓練與推論的核心模型。


````python
class ESGRoBERTaDataset(Dataset):
    """
    給 RoBERTa 使用的 ESG Dataset。

    每筆資料會輸出：
    1. input_ids
    2. attention_mask
    3. labels：若資料含 true label，會輸出 promise_status 和 evidence_status
    4. id
    5. raw text
    """

    def __init__(
        self,
        data,
        tokenizer,
        max_len,
        task_fields,
        label2id,
        text_field="input_text"
    ):
        self.data = data
        self.tokenizer = tokenizer
        self.max_len = max_len
        self.task_fields = task_fields
        self.label2id = label2id
        self.text_field = text_field

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        item = self.data[idx]

        text = item[self.text_field]

        encoded = self.tokenizer(
            text,
            padding="max_length",
            truncation=True,
            max_length=self.max_len,
            return_tensors="pt"
        )

        input_ids = encoded["input_ids"].squeeze(0)
        attention_mask = encoded["attention_mask"].squeeze(0)

        output = {
            "id": item["id"],
            "text": text,
            "input_ids": input_ids,
            "attention_mask": attention_mask,
        }

        has_all_labels = all(task in item for task in self.task_fields)

        if has_all_labels:
            labels = {}

            for task in self.task_fields:
                label_text = item[task]
                labels[task] = torch.tensor(
                    self.label2id[task][label_text],
                    dtype=torch.long
                )

            output["labels"] = labels

        return output

class MultiTaskRoBERTa(nn.Module):
    """
    多任務 RoBERTa 模型。

    共用同一個 RoBERTa encoder，
    但每個任務有自己的 classification head。

    本最終版 RoBERTa 負責：
    1. promise_status
    2. evidence_status
    """

    def __init__(
        self,
        model_name,
        task_fields,
        label_lists,
        dropout_rate=0.1
    ):
        super().__init__()

        self.model_name = model_name
        self.task_fields = task_fields
        self.label_lists = label_lists

        self.roberta = AutoModel.from_pretrained(model_name)

        if hasattr(self.roberta, "gradient_checkpointing_enable"):
            self.roberta.gradient_checkpointing_enable()
            print("✅ gradient checkpointing 已啟用")

        if hasattr(self.roberta.config, "use_cache"):
            self.roberta.config.use_cache = False

        hidden_size = self.roberta.config.hidden_size
        self.dropout = nn.Dropout(dropout_rate)

        self.classifiers = nn.ModuleDict()

        for task in task_fields:
            num_labels = len(label_lists[task])
            self.classifiers[task] = nn.Linear(hidden_size, num_labels)

        print("hidden_size:", hidden_size)
        print("分類頭：")
        for task in task_fields:
            print(f"- {task}: {len(label_lists[task])} labels")

    def forward(self, input_ids, attention_mask):
        outputs = self.roberta(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

        cls_output = outputs.last_hidden_state[:, 0, :]
        cls_output = self.dropout(cls_output)

        logits = {}

        for task in self.task_fields:
            logits[task] = self.classifiers[task](cls_output)

        return logits
````

### RoBERTa 訓練與評估函數

**執行類型：必跑。**

定義 FGM、指標計算、模型評估與單一 epoch 訓練流程。


````python
class FGM:
    """
    Fast Gradient Method 對抗訓練。

    簡單說：
    在 embedding 上加一點小擾動，讓模型不要只記住表面文字，
    可以提升模型泛化能力。
    """

    def __init__(self, model, emb_name="embeddings.word_embeddings", epsilon=1.0):
        self.model = model
        self.emb_name = emb_name
        self.epsilon = epsilon
        self.backup = {}

    def attack(self):
        for name, param in self.model.named_parameters():
            if param.requires_grad and self.emb_name in name:
                self.backup[name] = param.data.clone()

                norm = torch.norm(param.grad)

                if norm != 0 and not torch.isnan(norm):
                    r_at = self.epsilon * param.grad / norm
                    param.data.add_(r_at)

    def restore(self):
        for name, param in self.model.named_parameters():
            if param.requires_grad and self.emb_name in name:
                if name in self.backup:
                    param.data = self.backup[name]

        self.backup = {}

def compute_task_metrics(y_true, y_pred, labels):
    """
    計算單一任務的 macro F1、micro F1、accuracy。
    """

    macro_f1 = f1_score(
        y_true,
        y_pred,
        labels=labels,
        average="macro",
        zero_division=0
    )

    micro_f1 = f1_score(
        y_true,
        y_pred,
        labels=labels,
        average="micro",
        zero_division=0
    )

    acc = accuracy_score(y_true, y_pred)

    return {
        "macro_f1": macro_f1,
        "micro_f1": micro_f1,
        "accuracy": acc
    }

def compute_roberta_weighted_score(all_true_labels, all_pred_labels):
    """
    RoBERTa 目前只做：
    - promise_status
    - evidence_status

    這裡先用官方權重計算 Task 1 + Task 3 的部分分數。
    """

    task_results = {}
    weighted_score = 0.0

    for task in ROBERTA_TASK_FIELDS:
        labels = list(range(len(LABEL_LISTS[task])))

        metrics = compute_task_metrics(
            y_true=all_true_labels[task],
            y_pred=all_pred_labels[task],
            labels=labels
        )

        task_results[task] = metrics

        weighted_score += metrics["macro_f1"] * OFFICIAL_WEIGHTS[task]

    return weighted_score, task_results

def evaluate_model(model, data_loader, device, criterion_dict=None, show_report=False):
    """
    對 valid 或 test 做評估。

    回傳：
    1. avg_loss
    2. weighted_score
    3. task_results
    4. all_true_labels
    5. all_pred_labels
    6. all_pred_probs
    """

    model.eval()

    total_loss = 0.0
    total_steps = 0

    all_true_labels = {
        task: [] for task in ROBERTA_TASK_FIELDS
    }

    all_pred_labels = {
        task: [] for task in ROBERTA_TASK_FIELDS
    }

    all_pred_probs = {
        task: [] for task in ROBERTA_TASK_FIELDS
    }

    all_ids = []
    all_texts = []

    with torch.no_grad():
        for batch in tqdm(data_loader, desc="Evaluating", leave=False):
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)

            labels = {
                task: batch["labels"][task].to(device)
                for task in ROBERTA_TASK_FIELDS
            }

            logits = model(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

            loss = 0.0

            for task in ROBERTA_TASK_FIELDS:
                task_loss = criterion_dict[task](
                    logits[task],
                    labels[task]
                )
                loss += task_loss

            total_loss += loss.item()
            total_steps += 1

            for task in ROBERTA_TASK_FIELDS:
                probs = torch.softmax(logits[task], dim=1)
                preds = torch.argmax(probs, dim=1)

                all_true_labels[task].extend(labels[task].detach().cpu().numpy().tolist())
                all_pred_labels[task].extend(preds.detach().cpu().numpy().tolist())
                all_pred_probs[task].extend(probs.detach().cpu().numpy().tolist())

            all_ids.extend(batch["id"])
            all_texts.extend(batch["text"])

    avg_loss = total_loss / max(total_steps, 1)

    weighted_score, task_results = compute_roberta_weighted_score(
        all_true_labels,
        all_pred_labels
    )

    if show_report:
        print("\n" + "=" * 80)
        print("RoBERTa Evaluation Report")
        print("=" * 80)

        print("Average Loss:", round(avg_loss, 6))
        print("RoBERTa Weighted Partial Score:", round(weighted_score, 6))

        for task in ROBERTA_TASK_FIELDS:
            print("\n" + "=" * 60)
            print(task)
            print("=" * 60)

            print("Macro F1:", round(task_results[task]["macro_f1"], 4))
            print("Micro F1:", round(task_results[task]["micro_f1"], 4))
            print("Accuracy:", round(task_results[task]["accuracy"], 4))

            target_names = LABEL_LISTS[task]

            print(
                classification_report(
                    all_true_labels[task],
                    all_pred_labels[task],
                    labels=list(range(len(target_names))),
                    target_names=target_names,
                    zero_division=0
                )
            )

    return {
        "avg_loss": avg_loss,
        "weighted_score": weighted_score,
        "task_results": task_results,
        "all_true_labels": all_true_labels,
        "all_pred_labels": all_pred_labels,
        "all_pred_probs": all_pred_probs,
        "all_ids": all_ids,
        "all_texts": all_texts
    }

def train_one_epoch(
    model,
    data_loader,
    optimizer,
    scheduler,
    criterion_dict,
    device,
    scaler=None,
    use_fp16=True,
    use_fgm=True,
    accum_steps=16
):
    """
    訓練一個 epoch。

    支援：
    - FP16
    - FGM
    - Gradient Accumulation
    """

    model.train()

    total_loss = 0.0
    total_steps = 0

    optimizer.zero_grad()

    fgm = FGM(model) if use_fgm else None

    progress_bar = tqdm(data_loader, desc="Training", leave=False)

    for step, batch in enumerate(progress_bar):
        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)

        labels = {
            task: batch["labels"][task].to(device)
            for task in ROBERTA_TASK_FIELDS
        }

        if use_fp16 and scaler is not None:
            with torch.cuda.amp.autocast():
                logits = model(
                    input_ids=input_ids,
                    attention_mask=attention_mask
                )

                loss = 0.0

                for task in ROBERTA_TASK_FIELDS:
                    task_loss = criterion_dict[task](
                        logits[task],
                        labels[task]
                    )
                    loss += task_loss

                loss = loss / accum_steps

            scaler.scale(loss).backward()

        else:
            logits = model(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

            loss = 0.0

            for task in ROBERTA_TASK_FIELDS:
                task_loss = criterion_dict[task](
                    logits[task],
                    labels[task]
                )
                loss += task_loss

            loss = loss / accum_steps
            loss.backward()

        if use_fgm:
            fgm.attack()

            if use_fp16 and scaler is not None:
                with torch.cuda.amp.autocast():
                    adv_logits = model(
                        input_ids=input_ids,
                        attention_mask=attention_mask
                    )

                    adv_loss = 0.0

                    for task in ROBERTA_TASK_FIELDS:
                        task_adv_loss = criterion_dict[task](
                            adv_logits[task],
                            labels[task]
                        )
                        adv_loss += task_adv_loss

                    adv_loss = adv_loss / accum_steps

                scaler.scale(adv_loss).backward()

            else:
                adv_logits = model(
                    input_ids=input_ids,
                    attention_mask=attention_mask
                )

                adv_loss = 0.0

                for task in ROBERTA_TASK_FIELDS:
                    task_adv_loss = criterion_dict[task](
                        adv_logits[task],
                        labels[task]
                    )
                    adv_loss += task_adv_loss

                adv_loss = adv_loss / accum_steps
                adv_loss.backward()

            fgm.restore()

        if (step + 1) % accum_steps == 0 or (step + 1) == len(data_loader):
            if use_fp16 and scaler is not None:
                scaler.unscale_(optimizer)
                torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)

                scaler.step(optimizer)
                scaler.update()

            else:
                torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
                optimizer.step()

            scheduler.step()
            optimizer.zero_grad()

        total_loss += loss.item() * accum_steps
        total_steps += 1

        progress_bar.set_postfix({
            "loss": round(total_loss / max(total_steps, 1), 4)
        })

    avg_train_loss = total_loss / max(total_steps, 1)

    return avg_train_loss
````

````python
def train_single_roberta_seed(current_seed, include_seed_in_history=True):
    set_seed(current_seed)

    print("=" * 80)
    print(f"開始訓練 RoBERTa 單一模型 | Seed = {current_seed}")
    print("=" * 80)
    print("目前超參數組合:", SELECTED_CONFIG_NAME)
    print("TRAIN_VERSION:", TRAIN_VERSION)
    print("MODEL_NAME:", MODEL_NAME)
    print("ROBERTA_TASK_FIELDS:", ROBERTA_TASK_FIELDS)
    print("EPOCHS:", EPOCHS)
    print("LR:", LR)
    print("USE_FP16:", USE_FP16)
    print("USE_FGM:", USE_FGM)

    model = MultiTaskRoBERTa(
        model_name=MODEL_NAME,
        task_fields=ROBERTA_TASK_FIELDS,
        label_lists=LABEL_LISTS,
        dropout_rate=DROPOUT_RATE
    ).to(device)
    print("\n✅ 模型建立完成")

    optimizer = AdamW(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
    print("✅ Optimizer 建立完成")

    num_update_steps_per_epoch = len(train_loader) // ACCUM_STEPS
    if len(train_loader) % ACCUM_STEPS != 0:
        num_update_steps_per_epoch += 1

    total_training_steps = num_update_steps_per_epoch * EPOCHS
    warmup_steps = int(total_training_steps * WARMUP_RATIO)
    scheduler = get_linear_schedule_with_warmup(
        optimizer,
        num_warmup_steps=warmup_steps,
        num_training_steps=total_training_steps
    )

    print("✅ Scheduler 建立完成")
    print("num_update_steps_per_epoch:", num_update_steps_per_epoch)
    print("total_training_steps:", total_training_steps)
    print("warmup_steps:", warmup_steps)

    if USE_FP16 and torch.cuda.is_available():
        scaler = GradScaler()
        print("✅ FP16 GradScaler 已啟用")
    else:
        scaler = None
        print("⚠️ 未啟用 FP16")

    single_model_save_path = os.path.join(
        ROBERTA_MODEL_DIR,
        f"best_roberta_seed_{current_seed}.pt"
    )
    single_history_save_path = os.path.join(
        ROBERTA_MODEL_DIR,
        f"training_history_seed_{current_seed}.json"
    )

    print("\n模型將儲存到：")
    print(single_model_save_path)

    best_valid_score = -1
    best_epoch = -1
    training_history = []

    for epoch in range(1, EPOCHS + 1):
        print("\n" + "=" * 80)
        print(f"Seed {current_seed} | Epoch {epoch}/{EPOCHS}")
        print("=" * 80)

        train_loss = train_one_epoch(
            model=model,
            data_loader=train_loader,
            optimizer=optimizer,
            scheduler=scheduler,
            criterion_dict=criterion_dict,
            device=device,
            scaler=scaler,
            use_fp16=USE_FP16,
            use_fgm=USE_FGM,
            accum_steps=ACCUM_STEPS
        )

        valid_result = evaluate_model(
            model=model,
            data_loader=valid_loader,
            device=device,
            criterion_dict=criterion_dict,
            show_report=False
        )
        valid_score = valid_result["weighted_score"]
        valid_loss = valid_result["avg_loss"]

        print(f"Train Loss: {train_loss:.6f}")
        print(f"Valid Loss: {valid_loss:.6f}")
        print(f"Valid RoBERTa Partial Score: {valid_score:.6f}")

        for task in ROBERTA_TASK_FIELDS:
            task_metrics = valid_result["task_results"][task]
            print(
                f"{task} | "
                f"Macro F1: {task_metrics['macro_f1']:.4f} | "
                f"Micro F1: {task_metrics['micro_f1']:.4f} | "
                f"Acc: {task_metrics['accuracy']:.4f}"
            )

        epoch_record = {
            "epoch": epoch,
            "train_loss": train_loss,
            "valid_loss": valid_loss,
            "valid_roberta_partial_score": valid_score,
            "task_results": valid_result["task_results"]
        }
        if include_seed_in_history:
            epoch_record["seed"] = current_seed
        training_history.append(epoch_record)

        if valid_score > best_valid_score:
            best_valid_score = valid_score
            best_epoch = epoch
            torch.save(
                {
                    "model_state_dict": model.state_dict(),
                    "selected_config_name": SELECTED_CONFIG_NAME,
                    "seed": current_seed,
                    "epoch": epoch,
                    "best_valid_score": best_valid_score,
                    "model_name": MODEL_NAME,
                    "roberta_task_fields": ROBERTA_TASK_FIELDS,
                    "label_lists": LABEL_LISTS,
                    "label2id": label2id,
                    "id2label": id2label,
                    "hyperparameters": HP,
                    "train_version": TRAIN_VERSION
                },
                single_model_save_path
            )
            print(f"✨ Model saved! Epoch {epoch}, Score = {best_valid_score:.6f}")

        with open(single_history_save_path, "w", encoding="utf-8") as f:
            json.dump(training_history, f, ensure_ascii=False, indent=2)

    print("\n" + "=" * 80)
    print(" ")
    print("=" * 80)
    print("Seed:", current_seed)
    print("Best Epoch:", best_epoch)
    print("Best Valid RoBERTa Partial Score:", best_valid_score)
    print("Best Model Path:", single_model_save_path)
    print("Training History Path:", single_history_save_path)

    del model
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        print("✅ GPU cache 已清除")

    return {
        "seed": current_seed,
        "best_epoch": best_epoch,
        "best_valid_score": best_valid_score,
        "model_path": single_model_save_path,
        "history_path": single_history_save_path,
        "training_history": training_history
    }
````

### Ensemble 函數

**執行類型：必跑。**

載入 checkpoint、執行單模型預測，並將多個 seed 模型用 soft voting 整合。


````python
def make_json_serializable(obj):
    if isinstance(obj, torch.Tensor):
        return obj.detach().cpu().tolist()

    if isinstance(obj, np.ndarray):
        return obj.tolist()

    if isinstance(obj, np.integer):
        return int(obj)

    if isinstance(obj, np.floating):
        return float(obj)

    if isinstance(obj, np.bool_):
        return bool(obj)

    if isinstance(obj, dict):
        return {
            str(k): make_json_serializable(v)
            for k, v in obj.items()
        }

    if isinstance(obj, list):
        return [
            make_json_serializable(v)
            for v in obj
        ]

    if isinstance(obj, tuple):
        return [
            make_json_serializable(v)
            for v in obj
        ]

    return obj

def load_roberta_checkpoint_memory_safe(model_path, device):
    """
    Memory-safe 載入單一 RoBERTa checkpoint。
    一次只放一個模型到 GPU。
    """

    ckpt = torch.load(
        model_path,
        map_location="cpu"
    )

    model = MultiTaskRoBERTa(
        model_name=ckpt.get("model_name", MODEL_NAME),
        task_fields=ckpt.get("roberta_task_fields", ROBERTA_TASK_FIELDS),
        label_lists=ckpt.get("label_lists", LABEL_LISTS),
        dropout_rate=ckpt.get("hyperparameters", HP).get("DROPOUT_RATE", DROPOUT_RATE)
    )

    model.load_state_dict(ckpt["model_state_dict"])
    model.to(device)
    model.eval()

    print(
        f"✅ 已載入模型 | "
        f"seed={ckpt.get('seed')} | "
        f"epoch={ckpt.get('epoch')} | "
        f"valid_score={ckpt.get('best_valid_score')}"
    )

    return model, ckpt

def predict_probs_single_roberta_model(model, data_loader, device):
    """
    單一 RoBERTa 模型預測。
    回傳：
    - ids
    - texts
    - true labels
    - probs by task
    """

    model.eval()

    all_ids = []
    all_texts = []

    all_true_labels = {
        task: []
        for task in ROBERTA_TASK_FIELDS
    }

    all_probs = {
        task: []
        for task in ROBERTA_TASK_FIELDS
    }

    with torch.no_grad():
        for batch in tqdm(data_loader, desc="Single RoBERTa Predict"):
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)

            logits = model(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

            for task in ROBERTA_TASK_FIELDS:
                probs = torch.softmax(logits[task], dim=1)

                all_probs[task].append(
                    probs.detach().cpu().numpy()
                )

                if "labels" in batch:
                    labels = batch["labels"][task]
                    all_true_labels[task].extend(
                        labels.detach().cpu().numpy().tolist()
                    )

            all_ids.extend(batch["id"])
            all_texts.extend(batch["text"])

    all_probs_concat = {}

    for task in ROBERTA_TASK_FIELDS:
        all_probs_concat[task] = np.vstack(all_probs[task])

    return {
        "all_ids": all_ids,
        "all_texts": all_texts,
        "all_true_labels": all_true_labels,
        "all_probs": all_probs_concat
    }

def predict_roberta_ensemble_memory_safe(model_paths, data_loader, device):
    """
    Memory-safe soft voting:
    一次只載入一個模型，把 probs 累加到 CPU。
    """

    accumulated_probs = {
        task: None
        for task in ROBERTA_TASK_FIELDS
    }

    base_ids = None
    base_texts = None
    base_true_labels = None

    ckpt_summaries = []

    for model_idx, model_path in enumerate(model_paths):
        print("\n" + "=" * 100)
        print(f"載入第 {model_idx + 1}/{len(model_paths)} 個 RoBERTa 模型")
        print(model_path)
        print("=" * 100)

        model, ckpt = load_roberta_checkpoint_memory_safe(
            model_path=model_path,
            device=device
        )

        result = predict_probs_single_roberta_model(
            model=model,
            data_loader=data_loader,
            device=device
        )

        ckpt_summaries.append({
            "seed": ckpt.get("seed"),
            "epoch": ckpt.get("epoch"),
            "best_valid_score": ckpt.get("best_valid_score"),
            "model_path": model_path
        })

        if base_ids is None:
            base_ids = result["all_ids"]
            base_texts = result["all_texts"]
            base_true_labels = result["all_true_labels"]
            if all(len(v) == 0 for v in base_true_labels.values()):
                base_true_labels = None
        else:
            if [str(x) for x in base_ids] != [str(x) for x in result["all_ids"]]:
                raise ValueError("❌ 不同 seed 模型的預測 id 順序不一致")

        for task in ROBERTA_TASK_FIELDS:
            if accumulated_probs[task] is None:
                accumulated_probs[task] = result["all_probs"][task]
            else:
                accumulated_probs[task] += result["all_probs"][task]

        del model
        del ckpt
        del result

        gc.collect()

        if torch.cuda.is_available():
            torch.cuda.empty_cache()
            torch.cuda.ipc_collect()

        print("✅ 已釋放此模型 GPU memory")

    avg_probs = {
        task: accumulated_probs[task] / len(model_paths)
        for task in ROBERTA_TASK_FIELDS
    }

    pred_labels = {}

    for task in ROBERTA_TASK_FIELDS:
        pred_labels[task] = np.argmax(
            avg_probs[task],
            axis=1
        ).tolist()

    return {
        "all_ids": base_ids,
        "all_texts": base_texts,
        "all_true_labels": base_true_labels,
        "all_pred_labels": pred_labels,
        "all_pred_probs": {
            task: avg_probs[task].tolist()
            for task in ROBERTA_TASK_FIELDS
        },
        "ckpt_summaries": ckpt_summaries
    }
````

### RAG 檢索函數

**執行類型：必跑。**

負責文字向量化、相似範例檢索，以及簡單 retrieval majority baseline 評估。


````python
def average_pool(last_hidden_states, attention_mask):
    """
    multilingual-e5 官方常用的 average pooling。
    只平均有效 token，不平均 padding。
    """
    last_hidden = last_hidden_states.masked_fill(
        ~attention_mask[..., None].bool(),
        0.0
    )

    return last_hidden.sum(dim=1) / attention_mask.sum(dim=1)[..., None]

def encode_texts_e5(texts, batch_size=16, max_length=512):
    """
    把一批文字轉成 multilingual-e5-base embedding。

    注意：
    e5 模型建議格式：
    - retrieval 文件：passage: 文字
    - 查詢文字：query: 文字
    """

    all_embeddings = []

    for start in tqdm(range(0, len(texts), batch_size), desc="Encoding texts"):
        batch_texts = texts[start:start + batch_size]

        encoded = embedding_tokenizer(
            batch_texts,
            padding=True,
            truncation=True,
            max_length=max_length,
            return_tensors="pt"
        ).to(device)

        with torch.no_grad():
            outputs = embedding_model(**encoded)

            embeddings = average_pool(
                outputs.last_hidden_state,
                encoded["attention_mask"]
            )

            embeddings = F.normalize(embeddings, p=2, dim=1)

        all_embeddings.append(
            embeddings.detach().cpu().numpy()
        )

    all_embeddings = np.vstack(all_embeddings)

    return all_embeddings

def retrieve_top_k_examples(query_text, top_k=6):
    """
    給一段 query_text，從 retrieval_pool 裡找 top-k 相似案例。
    """

    query_input = ["query: " + query_text]

    query_embedding = encode_texts_e5(
        texts=query_input,
        batch_size=1,
        max_length=512
    )

    similarities = np.matmul(
        retrieval_embeddings,
        query_embedding[0]
    )

    top_indices = similarities.argsort()[::-1][:top_k]

    top_examples = []

    for rank, idx in enumerate(top_indices, start=1):
        item = retrieval_pool_with_index[idx].copy()

        item["rank"] = rank
        item["similarity"] = float(similarities[idx])

        top_examples.append(item)

    return top_examples

def retrieve_top_k_examples_no_self(query_text, query_id=None, top_k=6):
    """
    給一段 query_text，從 retrieval_pool 裡找 top-k 相似案例。
    如果 query_id 出現在 retrieval_pool 裡，就排除自己。
    """

    query_input = ["query: " + query_text]

    query_embedding = encode_texts_e5(
        texts=query_input,
        batch_size=1,
        max_length=512
    )

    similarities = np.matmul(
        retrieval_embeddings,
        query_embedding[0]
    )

    sorted_indices = similarities.argsort()[::-1]

    top_examples = []

    for idx in sorted_indices:
        item = retrieval_pool_with_index[idx].copy()

        if query_id is not None and item["id"] == query_id:
            continue

        item["rank"] = len(top_examples) + 1
        item["similarity"] = float(similarities[idx])

        top_examples.append(item)

        if len(top_examples) >= top_k:
            break

    return top_examples

def majority_vote(labels):
    """
    對 top-k retrieved examples 的標籤做多數決。
    若平手，選第一個出現的標籤。
    """
    counter = Counter(labels)
    max_count = max(counter.values())

    candidates = [
        label for label, count in counter.items()
        if count == max_count
    ]

    for label in labels:
        if label in candidates:
            return label

def retrieval_majority_baseline_no_self(eval_data, top_k):
    """
    不使用 GPT，只用 top-k 檢索案例的標籤多數決。
    這版會排除 query 自己。
    """

    predictions = []

    for item in tqdm(eval_data, desc=f"No-self retrieval baseline top_k={top_k}"):
        query_text = item["data"]
        query_id = item["id"]

        top_examples = retrieve_top_k_examples_no_self(
            query_text=query_text,
            query_id=query_id,
            top_k=top_k
        )

        pred_item = {
            "id": item["id"],
            "data": item["data"],
            "top_k": top_k,
            "retrieved_ids": [ex["id"] for ex in top_examples],
            "retrieved_similarities": [ex["similarity"] for ex in top_examples]
        }

        for task in RAG_TASK_FIELDS:
            retrieved_labels = [
                ex[task] for ex in top_examples
            ]

            pred_label = majority_vote(retrieved_labels)

            pred_item[f"{task}_true"] = item[task]
            pred_item[f"{task}_pred"] = pred_label
            pred_item[f"{task}_retrieved_labels"] = retrieved_labels

        predictions.append(pred_item)

    return predictions
````

### GPT 呼叫、標籤正規化與共用預測函數

**執行類型：必跑。**

集中處理 GPT JSON 解析、API 呼叫、標籤正規化，並用一個共用函數支援 Task 2 與 Task 4 的 v4 預測。


````python
def safe_parse_json(text):
    """
    安全解析 GPT 輸出的 JSON，可處理 code block 或前後多餘文字。
    """
    if text is None:
        raise ValueError("GPT 回傳內容是 None")

    text = str(text).strip()
    text = text.replace("```json", "")
    text = text.replace("```JSON", "")
    text = text.replace("```", "")
    text = text.strip()

    try:
        return json.loads(text)
    except Exception:
        pass

    match = re.search(r"\{.*\}", text, re.DOTALL)
    if match:
        json_text = match.group(0)
        try:
            return json.loads(json_text)
        except Exception as e:
            raise ValueError(f"找到 JSON 區塊但解析失敗：{json_text}") from e

    raise ValueError(f"無法從 GPT 輸出解析 JSON：{text}")


def call_gpt_json(prompt, model=None, temperature=None):
    """
    呼叫 GPT 並回傳 parsed JSON 與 raw text。
    gpt-5 系列不傳 temperature；其他模型才傳 temperature。
    """
    if model is None:
        model = GPT_MODEL
    if temperature is None:
        temperature = GPT_TEMPERATURE

    last_error = None
    for attempt in range(1, GPT_MAX_RETRIES + 1):
        try:
            request_kwargs = {
                "model": model,
                "messages": [
                    {
                        "role": "system",
                        "content": "你是一位嚴謹的 ESG 文本標註專家。請嚴格依照使用者要求，只輸出合法 JSON。",
                    },
                    {
                        "role": "user",
                        "content": prompt,
                    },
                ],
            }

            if not str(model).startswith("gpt-5"):
                request_kwargs["temperature"] = temperature

            response = client.chat.completions.create(**request_kwargs)
            raw_text = response.choices[0].message.content
            parsed = safe_parse_json(raw_text)
            return parsed, raw_text

        except Exception as e:
            last_error = e
            print(f"⚠️ GPT 呼叫或解析失敗，第 {attempt}/{GPT_MAX_RETRIES} 次")
            print("model:", model)
            print("錯誤訊息:", e)
            time.sleep(GPT_SLEEP_SECONDS * attempt)

    raise RuntimeError(f"GPT 呼叫失敗，已重試 {GPT_MAX_RETRIES} 次") from last_error


def normalize_verification_timeline(x):
    if x is None:
        return "N/A"

    s = str(x).strip().lower()
    mapping = {
        "already": "already",
        "already verified": "already",
        "verified already": "already",
        "within_2_years": "within_2_years",
        "within 2 years": "within_2_years",
        "within two years": "within_2_years",
        "less than 2 years": "within_2_years",
        "<2 years": "within_2_years",
        "between_2_and_5_years": "between_2_and_5_years",
        "between 2 and 5 years": "between_2_and_5_years",
        "between two and five years": "between_2_and_5_years",
        "2-5 years": "between_2_and_5_years",
        "2 to 5 years": "between_2_and_5_years",
        "longer_than_5_years": "longer_than_5_years",
        "longer than 5 years": "longer_than_5_years",
        "more than 5 years": "longer_than_5_years",
        ">5 years": "longer_than_5_years",
        "n/a": "N/A",
        "na": "N/A",
        "none": "N/A",
        "null": "N/A",
        "": "N/A",
        "not applicable": "N/A",
    }
    return mapping.get(s, str(x).strip())


def normalize_evidence_quality(x):
    if x is None:
        return "N/A"

    s = str(x).strip().lower()
    mapping = {
        "clear": "Clear",
        "clearly": "Clear",
        "sufficient": "Clear",
        "high": "Clear",
        "not clear": "Not Clear",
        "not_clear": "Not Clear",
        "unclear": "Not Clear",
        "ambiguous": "Not Clear",
        "insufficient": "Not Clear",
        "misleading": "Misleading",
        "mislead": "Misleading",
        "n/a": "N/A",
        "na": "N/A",
        "none": "N/A",
        "null": "N/A",
        "": "N/A",
        "not applicable": "N/A",
    }
    return mapping.get(s, str(x).strip())


def predict_gpt_rag_task(
    test_item,
    task_name,
    label_key,
    prompt_builder,
    normalize_func,
    top_k=None,
):
    """
    共用 GPT + RAG 預測流程。

    Task 2 / Task 4 的差異只在 prompt_builder、label_key 與 normalize_func。
    """
    if top_k is None:
        top_k = BEST_TOP_K

    item_id = normalize_id(test_item["id"])
    test_text = test_item["data"]
    retrieved_examples = retrieve_top_k_examples_no_self(
        query_text=test_text,
        query_id=item_id,
        top_k=top_k,
    )
    roberta_pred = roberta_pred_by_id[item_id]
    prompt = prompt_builder(
        test_text=test_text,
        retrieved_examples=retrieved_examples,
        roberta_pred=roberta_pred,
    )
    parsed, raw_text = call_gpt_json(prompt=prompt, model=GPT_MODEL)
    pred_label = normalize_func(parsed.get(label_key, "N/A"))

    return {
        "id": item_id,
        "task": task_name,
        "prediction": pred_label,
        "raw_response": raw_text,
        "parsed_response": parsed,
        "retrieved_ids": [normalize_id(ex["id"]) for ex in retrieved_examples],
        "retrieved_similarities": [float(ex["similarity"]) for ex in retrieved_examples],
        "prompt_length": len(prompt),
    }


def predict_timeline_with_gpt_rag(test_item, top_k=None):
    return predict_gpt_rag_task(
        test_item=test_item,
        task_name="verification_timeline",
        label_key="verification_timeline",
        prompt_builder=build_timeline_prompt_v4,
        normalize_func=normalize_verification_timeline,
        top_k=top_k,
    )


def predict_quality_with_gpt_rag(test_item, top_k=None):
    return predict_gpt_rag_task(
        test_item=test_item,
        task_name="evidence_quality",
        label_key="evidence_quality",
        prompt_builder=build_quality_prompt_v4,
        normalize_func=normalize_evidence_quality,
        top_k=top_k,
    )
````

### Prompt v4 建立函數

**執行類型：必跑。**

````python
def build_timeline_prompt_v4(test_text, retrieved_examples, roberta_pred):
    """
    Timeline prompt v4:
    基於 v3，加強：
    1. 不要過度判 already
    2. 背景/制度/風險描述優先考慮 N/A
    3. 未來行動/持續推動優先考慮 between_2_and_5_years
    """

    examples_text = ""

    for i, ex in enumerate(retrieved_examples, 1):
        ex_text = ex.get("retrieval_text", ex.get("text", ""))

        examples_text += f"""
【相似範例 {i}】
文本：
{ex_text}

標籤：
verification_timeline = {ex["verification_timeline"]}
promise_status = {ex["promise_status"]}
evidence_status = {ex["evidence_status"]}
"""

    prompt = f"""
你是一位嚴謹的 ESG Promise Verification 標註專家。
請根據測試文本、RoBERTa 輔助預測、標籤定義與相似範例，只判斷 verification_timeline。

你只需要預測：
1. verification_timeline

請不要預測 promise_status。
請不要預測 evidence_status。
請不要預測 evidence_quality。
請只輸出 JSON。

============================================================
RoBERTa 輔助預測
============================================================

promise_status = {roberta_pred["promise_status"]}
evidence_status = {roberta_pred["evidence_status"]}

注意：
- RoBERTa 預測只是輔助，不一定完全正確。
- verification_timeline 主要看「承諾或行動的時間狀態」。
- 不要只因為文本出現年份、數字、制度、政策名稱，就自動判 already。
- 要判斷文本重點是在「已完成成果」、「未來承諾」，還是「背景/制度說明」。

============================================================
verification_timeline 標籤定義
============================================================

- already：
文本主要描述 ESG 行動、成果、制度、活動、認證、揭露、調查、稽核或數據已經發生、已經存在、已經完成。
常見線索：
「已完成」、「已取得」、「已通過」、「已達成」、「已執行」、「已設立」、
「於 2024 年完成」、「2024 年共...」、「累計...」、「已累積...」、「自某年起已持續...」。
注意：already 必須有明確已完成的 ESG 行動、成果、驗證、認證、活動或量化結果。

- within_2_years：
文本明確表示某項承諾、目標或行動會在 2 年內完成。
只有明確短期期限時才使用，例如：
「一年內」、「兩年內」、「2025 年前」、「2026 年前」。
如果沒有明確短期完成期限，不要亂判 within_2_years。

- between_2_and_5_years：
文本描述未來承諾、後續追蹤、逐步推動、持續改善、持續優化、階段性推動，且不像是超長期目標。
常見線索：
「將持續」、「後續追蹤」、「持續改善」、「持續優化」、「持續精進」、
「逐步建立」、「逐步推動」、「規劃」、「訂定目標」、「階段性推動」、
「期待」、「期盼」、「共同推動」、「未來將...」。

- longer_than_5_years：
文本明確描述長期承諾或長期目標。
常見線索：
「長期目標」、「中長期目標」、「短中長期目標」、
「2030」、「2040」、「2050」、「淨零」、「碳中和」。

- N/A：
文本沒有明確承諾時間，或只是描述背景、議題重要性、治理架構、組織職責、管理流程、政策原則、風險評估方法。
常見線索：
「委員會」、「執行小組」、「治理架構」、「管理架構」、「政策制定」、
「統籌推動」、「負責執行」、「最高督導單位」、「風險與機會」、「評估標準」、「議題說明」。

============================================================
重要判斷規則 v4
============================================================

1. 先判斷文本是否真的有「可標註的時間狀態」：
   - 如果只是背景介紹、議題說明、制度架構、風險評估、管理流程，沒有明確承諾或成果時間，優先判 N/A。

2. 不要過度判 already：
   - 有年份或數字，不代表一定是 already。
   - already 只用在文本明確描述「已完成的 ESG 行動、成果、認證、活動、揭露或量化結果」。

3. 如果文本同時有「已完成事實」和「未來持續追蹤」：
   - 若文本重點是未來持續改善、逐步推動、後續追蹤，偏 between_2_and_5_years。
   - 若未來只是附帶說明，主體是已完成成果，才判 already。

4. 如果出現「將、持續、推動、規劃、逐步、後續、未來、目標」：
   - 不要直接判 already。
   - 若不是 2030/2050 等長期目標，優先考慮 between_2_and_5_years。

5. 「2030 / 2040 / 2050 / 淨零 / 碳中和」通常表示 longer_than_5_years。
6. 沒有明確時間或承諾時間時，不要硬猜，判 N/A。

============================================================
相似範例
============================================================

{examples_text}

============================================================
測試文本
============================================================

{test_text}

============================================================
輸出格式
============================================================

請只輸出以下 JSON 格式，不要加入 markdown，不要加入說明文字：

{{
  "verification_timeline": "already"
}}
"""

    return prompt.strip()

def build_quality_prompt_v4(test_text, retrieved_examples, roberta_pred):
    """
    Quality prompt v4:
    基於 v3，加強：
    1. 先判斷是否適用 evidence_quality
    2. 減少 N/A 被誤判成 Clear / Not Clear
    3. Clear 必須直接支持 ESG 承諾或成果
    """

    examples_text = ""

    for i, ex in enumerate(retrieved_examples, 1):
        ex_text = ex.get("retrieval_text", ex.get("text", ""))

        examples_text += f"""
【相似範例 {i}】
文本：
{ex_text}

標籤：
promise_status = {ex["promise_status"]}
evidence_status = {ex["evidence_status"]}
evidence_quality = {ex["evidence_quality"]}
"""

    prompt = f"""
你是一位嚴謹的 ESG Promise Verification 標註專家。
請根據測試文本、RoBERTa 輔助預測、標籤定義與相似範例，只判斷 evidence_quality。

你只需要預測：
1. evidence_quality

請不要預測 verification_timeline。
請不要預測 promise_status。
請不要預測 evidence_status。
請只輸出 JSON。

============================================================
RoBERTa 輔助預測
============================================================

promise_status = {roberta_pred["promise_status"]}
evidence_status = {roberta_pred["evidence_status"]}

注意：
- RoBERTa 預測只是輔助，不一定完全正確。
- evidence_quality 必須先判斷「是否適用」。
- 不要只因為文本出現數字、年份、ISO、政策名稱，就自動判 Clear。
- Clear 必須是該證據直接支持 ESG 承諾、行動或成果。
- 如果文本只是背景介紹、議題說明、制度架構、政策原則、風險評估，通常應判 N/A。

============================================================
evidence_quality 標籤定義
============================================================

- Clear：
文本提供具體、可驗證，且直接支持 ESG 承諾或 ESG 成果的證據。
常見證據包括：
明確年份、金額、百分比、數量、人次、件數、場次、第三方驗證、ISO 認證、SBTi、獎項、具體成果、量化指標、已完成成果、達成率、改善率、投入金額、捐助金額。
注意：數字或認證必須和 ESG 行動/成果直接相關，才算 Clear。

- Not Clear：
文本有 ESG 承諾、方向、政策、制度、行動或管理描述，但證據不足、不夠具體。
常見情況：
只有「致力於」、「持續推動」、「強化」、「提升」、「落實」、「重視」、「改善」、「遵循」等方向性語句，
但缺少可驗證數據、具體成果、明確認證、量化指標。

- Misleading：
文本使用很正式、正面、承諾式、治理式或社會責任語氣，使讀者可能誤以為公司已有明確 ESG 作為、證據或驗證，
但實際上沒有具體數據、沒有驗證依據、沒有明確成果、沒有可查證證據。
Misleading 比 Not Clear 更像「過度包裝」或「容易讓人誤解已有證據」。
注意：如果文本其實有明確量化證據或認證，不要判 Misleading。

- N/A：
文本不適用 evidence_quality 判斷。
常見情況：
沒有 ESG promise、沒有 evidence、只是背景介紹、議題說明、產業趨勢、組織架構、管理職責、風險評估、政策描述，且不構成可評估的證據品質。

============================================================
重要判斷規則 v4
============================================================

1. 先判斷是否適用 evidence_quality：
   - 如果文本只是背景介紹、議題說明、風險描述、制度架構、管理流程、政策原則，沒有明確 ESG 承諾或證據，優先判 N/A。

2. 如果 promise_status 或 evidence_status 被 RoBERTa 判成 No / N/A：
   - 除非文本本身明確有 ESG 承諾與證據，否則 evidence_quality 優先判 N/A。

3. 不要過度判 Clear：
   - 不是看到數字、年份、ISO、政策名稱就判 Clear。
   - Clear 必須是證據直接支持 ESG 承諾、成果或已完成行動。

4. 不要過度判 Not Clear：
   - 如果文本根本不適用 evidence_quality，不要判 Not Clear，應判 N/A。
   - Not Clear 用於「有承諾/行動方向，但證據不足」的情境。

5. Misleading 只在文本明顯有過度正面包裝、承諾式語氣，且缺乏具體證據時使用。
   - 如果文本有明確量化證據或具體認證，不要判 Misleading。
   - 如果只是一般證據不足，判 Not Clear，不要過度判 Misleading。

6. 如果 evidence_status = Yes：
   - 有具體、直接支持 ESG 承諾或成果的數據/認證/成果 → Clear
   - 有 ESG 行動方向但缺乏具體可驗證證據 → Not Clear
   - 強烈 ESG 包裝但沒有證據 → Misleading

============================================================
相似範例
============================================================

{examples_text}

============================================================
測試文本
============================================================

{test_text}

============================================================
輸出格式
============================================================

請只輸出以下 JSON 格式，不要加入 markdown，不要加入說明文字：

{{
  "evidence_quality": "Clear"
}}
"""

    return prompt.strip()
````

### 最終評分函數

**執行類型：必跑。**

根據四個任務的預測結果計算官方加權分數，供 hybrid 與 post-processing 評估共用。


````python
def compute_final_score_from_predictions(predictions, title="Evaluation"):
    """
    給 final predictions，計算四任務官方加權分數。
    """

    eval_results = {}
    total_score = 0.0

    print("\n" + "=" * 100)
    print(title)
    print("=" * 100)

    for task in FINAL_TASK_FIELDS:
        y_true = [
            item[f"{task}_true"]
            for item in predictions
        ]

        y_pred = [
            item[f"{task}_pred"]
            for item in predictions
        ]

        labels = LABEL_LISTS[task]

        macro_f1 = f1_score(
            y_true,
            y_pred,
            labels=labels,
            average="macro",
            zero_division=0
        )

        micro_f1 = f1_score(
            y_true,
            y_pred,
            labels=labels,
            average="micro",
            zero_division=0
        )

        acc = accuracy_score(y_true, y_pred)

        weighted_score = macro_f1 * OFFICIAL_WEIGHTS[task]
        total_score += weighted_score

        eval_results[task] = {
            "macro_f1": macro_f1,
            "micro_f1": micro_f1,
            "accuracy": acc,
            "weight": OFFICIAL_WEIGHTS[task],
            "weighted_score": weighted_score
        }

        print("\n" + "-" * 80)
        print(task)
        print("-" * 80)
        print("Macro F1:", round(macro_f1, 4))
        print("Micro F1:", round(micro_f1, 4))
        print("Accuracy:", round(acc, 4))
        print("Weighted Score:", round(weighted_score, 6))

    print("\n" + "=" * 100)
    print("Total Official Weighted Score:", round(total_score, 6))
    print("=" * 100)

    return total_score, eval_results


def apply_task_dependency_rules(prediction):
    """
    套用四任務前後關係規則，只修正預測欄位。

    規則：
    1. promise_status_pred == "No" 時，
       verification_timeline_pred / evidence_status_pred / evidence_quality_pred 都改為 "N/A"。
    2. evidence_status_pred == "No" 時，evidence_quality_pred 改為 "N/A"。
    """
    item = copy.deepcopy(prediction)
    changes = []

    if item.get("promise_status_pred") == "No":
        for field in [
            "verification_timeline_pred",
            "evidence_status_pred",
            "evidence_quality_pred",
        ]:
            if item.get(field) != "N/A":
                changes.append({
                    "field": field,
                    "before": item.get(field),
                    "after": "N/A",
                    "reason": "promise_status_pred == No",
                })
                item[field] = "N/A"

    if item.get("evidence_status_pred") == "No":
        field = "evidence_quality_pred"
        if item.get(field) != "N/A":
            changes.append({
                "field": field,
                "before": item.get(field),
                "after": "N/A",
                "reason": "evidence_status_pred == No",
            })
            item[field] = "N/A"

    item["dependency_rule_changes"] = changes
    item["dependency_rule_applied"] = len(changes) > 0
    return item

````

### 檢查資料欄位

**執行類型：必跑。**

確認必要欄位、文字內容、標籤完整性與 id 是否重複，並建立資料預覽。


````python
REQUIRED_FIELDS = [
    "id",
    "data",
    "promise_status",
    "verification_timeline",
    "evidence_status",
    "evidence_quality"
]

TASK_FIELDS = [
    "promise_status",
    "verification_timeline",
    "evidence_status",
    "evidence_quality"
]

print("必要欄位：")
print(REQUIRED_FIELDS)

not_dict_items = []

for idx, item in enumerate(raw_data):
    if not isinstance(item, dict):
        not_dict_items.append(idx)

if len(not_dict_items) > 0:
    raise TypeError(f"❌ 有資料不是 dict，位置：{not_dict_items[:10]}")
else:
    print("\n✅ 每筆資料都是 dict")

missing_field_records = []

for idx, item in enumerate(raw_data):
    missing_fields = [field for field in REQUIRED_FIELDS if field not in item]

    if len(missing_fields) > 0:
        missing_field_records.append({
            "index": idx,
            "id": item.get("id", None),
            "missing_fields": missing_fields
        })

if len(missing_field_records) > 0:
    print("\n❌ 發現缺少必要欄位的資料：")
    display(pd.DataFrame(missing_field_records).head(20))
    raise ValueError("資料有缺少必要欄位，請先修正")
else:
    print("✅ 所有資料都有必要欄位")

bad_text_records = []

for idx, item in enumerate(raw_data):
    text = item.get("data", None)

    if not isinstance(text, str) or len(text.strip()) == 0:
        bad_text_records.append({
            "index": idx,
            "id": item.get("id", None),
            "data": text
        })

if len(bad_text_records) > 0:
    print("\n❌ 發現 data 欄位不是有效文字：")
    display(pd.DataFrame(bad_text_records).head(20))
    raise ValueError("data 欄位有問題，請先修正")
else:
    print("✅ 所有 data 欄位都是有效文字")

missing_label_records = []

for idx, item in enumerate(raw_data):
    missing_labels = []

    for field in TASK_FIELDS:
        value = item.get(field, None)

        if value is None or str(value).strip() == "":
            missing_labels.append(field)

    if len(missing_labels) > 0:
        missing_label_records.append({
            "index": idx,
            "id": item.get("id", None),
            "missing_labels": missing_labels
        })

if len(missing_label_records) > 0:
    print("\n❌ 發現任務標籤缺失：")
    display(pd.DataFrame(missing_label_records).head(20))
    raise ValueError("有任務標籤缺失，請先修正")
else:
    print("✅ 四個任務標籤都沒有缺失")

ids = [item["id"] for item in raw_data]
id_counter = Counter(ids)

duplicated_ids = [
    item_id for item_id, count in id_counter.items()
    if count > 1
]

if len(duplicated_ids) > 0:
    print("\n❌ 發現重複 id：")
    print(duplicated_ids[:20])
    raise ValueError("資料中有重複 id，請先修正")
else:
    print("✅ id 沒有重複")

all_keys = set()

for item in raw_data:
    all_keys.update(item.keys())

print("\n資料中出現過的所有欄位：")
for key in sorted(all_keys):
    print("-", key)

preview_df = pd.DataFrame(raw_data)

print("\n資料預覽：")
display(preview_df[REQUIRED_FIELDS].head())

print("\n✅ Cell 3 完成：必要欄位與資料格式檢查通過")
````

### 標籤與官方權重

**執行類型：必跑。**

定義四個任務的合法標籤、label id 對應表，以及官方加權分數設定。


````python
TASK_FIELDS = [
    "promise_status",
    "verification_timeline",
    "evidence_status",
    "evidence_quality"
]

print("四個任務欄位：")
for task in TASK_FIELDS:
    print("-", task)

LABEL_LISTS = {
    "promise_status": [
        "Yes",
        "No"
    ],

    "verification_timeline": [
        "already",
        "within_2_years",
        "between_2_and_5_years",
        "longer_than_5_years",
        "N/A"
    ],

    "evidence_status": [
        "Yes",
        "No",
        "N/A"
    ],

    "evidence_quality": [
        "Clear",
        "Not Clear",
        "Misleading",
        "N/A"
    ]
}

print("\n各任務合法標籤：")
for task, labels in LABEL_LISTS.items():
    print(f"{task}: {labels}")

label2id = {}
id2label = {}

for task, labels in LABEL_LISTS.items():
    label2id[task] = {
        label: idx for idx, label in enumerate(labels)
    }

    id2label[task] = {
        idx: label for idx, label in enumerate(labels)
    }

print("\nlabel2id：")
for task in TASK_FIELDS:
    print(task, "=>", label2id[task])

print("\nid2label：")
for task in TASK_FIELDS:
    print(task, "=>", id2label[task])

OFFICIAL_WEIGHTS = {
    "promise_status": 0.20,
    "verification_timeline": 0.15,
    "evidence_status": 0.30,
    "evidence_quality": 0.35
}

print("\n官方計分權重：")
for task, weight in OFFICIAL_WEIGHTS.items():
    print(f"{task}: {weight}")

print("權重總和:", sum(OFFICIAL_WEIGHTS.values()))

if abs(sum(OFFICIAL_WEIGHTS.values()) - 1.0) > 1e-8:
    raise ValueError("❌ 官方權重總和不是 1，請檢查 OFFICIAL_WEIGHTS")
else:
    print("✅ 官方權重總和正確")

invalid_label_records = []

for idx, item in enumerate(raw_data):
    for task in TASK_FIELDS:
        value = item.get(task)

        if value not in LABEL_LISTS[task]:
            invalid_label_records.append({
                "index": idx,
                "id": item.get("id", None),
                "task": task,
                "invalid_value": value,
                "valid_labels": LABEL_LISTS[task]
            })

if len(invalid_label_records) > 0:
    print("\n❌ 發現不合法標籤：")
    display(pd.DataFrame(invalid_label_records).head(50))
    raise ValueError("資料中有不合法標籤，請先修正")
else:
    print("\n✅ 所有標籤都在合法範圍內")

print("\n各任務標籤分布：")

for task in TASK_FIELDS:
    counts = Counter([item[task] for item in raw_data])

    print("\n" + "=" * 60)
    print(task)
    print("=" * 60)

    total = len(raw_data)

    dist_df = pd.DataFrame([
        {
            "label": label,
            "count": counts.get(label, 0),
            "ratio": round(counts.get(label, 0) / total, 4)
        }
        for label in LABEL_LISTS[task]
    ])

    display(dist_df)

print("\n✅ Cell 4 完成：任務標籤、label mapping、官方權重都已設定完成")
````

### 類別分布檢查

**執行類型：建議跑；用來確認少數類別狀態。**

檢查各任務標籤分布，特別確認少數類別 `Misleading` 的樣本狀態。


````python
misleading_items = [
    item for item in raw_data
    if item["evidence_quality"] == "Misleading"
]

print("Misleading 筆數:", len(misleading_items))

if len(misleading_items) == 0:
    raise ValueError("❌ 找不到 Misleading 資料，請檢查資料版本")

if len(misleading_items) > 1:
    print("⚠️ Misleading 超過 1 筆，請確認是否符合預期")
else:
    print("✅ Misleading 只有 1 筆，符合目前已知資料狀況")

print("\nMisleading 資料：")
for item in misleading_items:
    print("=" * 80)
    print("id:", item["id"])
    print("promise_status:", item["promise_status"])
    print("verification_timeline:", item["verification_timeline"])
    print("evidence_status:", item["evidence_status"])
    print("evidence_quality:", item["evidence_quality"])
    print("\n原文 data:")
    print(item["data"])

misleading_df = pd.DataFrame(misleading_items)

display_cols = [
    "id",
    "promise_status",
    "verification_timeline",
    "evidence_status",
    "evidence_quality",
    "data"
]

print("\nMisleading 表格：")
display(misleading_df[display_cols])

print("\n四個任務標籤分布總表：")

distribution_rows = []

for task in TASK_FIELDS:
    counts = Counter([item[task] for item in raw_data])
    total = len(raw_data)

    for label in LABEL_LISTS[task]:
        distribution_rows.append({
            "task": task,
            "label": label,
            "count": counts.get(label, 0),
            "ratio": round(counts.get(label, 0) / total, 4)
        })

dist_df = pd.DataFrame(distribution_rows)

display(dist_df)

print("\n少數類別檢查：")

minority_threshold = 20

minority_df = dist_df[dist_df["count"] <= minority_threshold].copy()
minority_df = minority_df.sort_values(["task", "count"])

display(minority_df)

MISLEADING_IDS = [item["id"] for item in misleading_items]

print("\nMISLEADING_IDS:", MISLEADING_IDS)

print("\n✅ Cell 5 完成：已確認 Misleading 資料，下一格會固定把它放進 test set")
````

### 切分 train / valid / test

**執行類型：重建資料切分時必跑。**

建立訓練、驗證與測試集，並固定把稀有的 `Misleading` 樣本放入 test。


````python
TEST_SIZE = 200
VALID_RATIO_IN_DEVELOP = 0.125

SPLIT_SEED = 42

print("TEST_SIZE:", TEST_SIZE)
print("VALID_RATIO_IN_DEVELOP:", VALID_RATIO_IN_DEVELOP)
print("SPLIT_SEED:", SPLIT_SEED)

misleading_data = [
    copy.deepcopy(item) for item in raw_data
    if item["evidence_quality"] == "Misleading"
]

non_misleading_data = [
    copy.deepcopy(item) for item in raw_data
    if item["evidence_quality"] != "Misleading"
]

print("\nMisleading 筆數:", len(misleading_data))
print("非 Misleading 筆數:", len(non_misleading_data))

if len(misleading_data) == 0:
    raise ValueError("❌ 找不到 Misleading，不能執行固定 Misleading test split")

remaining_test_size = TEST_SIZE - len(misleading_data)

print("還需要從非 Misleading 抽出的 test 筆數:", remaining_test_size)

non_misleading_labels = [
    item["evidence_quality"] for item in non_misleading_data
]

non_mislead_train_dev, non_mislead_test = train_test_split(
    non_misleading_data,
    test_size=remaining_test_size,
    random_state=SPLIT_SEED,
    stratify=non_misleading_labels
)

test_data = misleading_data + non_mislead_test

develop_data = non_mislead_train_dev

print("\n切分後資料量：")
print("develop_data:", len(develop_data))
print("test_data:", len(test_data))

if len(test_data) != TEST_SIZE:
    raise ValueError(f"❌ test_data 筆數不是 {TEST_SIZE}，目前是 {len(test_data)}")

if len(develop_data) + len(test_data) != len(raw_data):
    raise ValueError("❌ develop_data + test_data 筆數不等於 raw_data")

develop_labels = [
    item["evidence_quality"] for item in develop_data
]

train_data, valid_data = train_test_split(
    develop_data,
    test_size=VALID_RATIO_IN_DEVELOP,
    random_state=SPLIT_SEED,
    stratify=develop_labels
)

print("\nTrain / Valid / Test 筆數：")
print("train_data:", len(train_data))
print("valid_data:", len(valid_data))
print("test_data:", len(test_data))

train_ids = set(item["id"] for item in train_data)
valid_ids = set(item["id"] for item in valid_data)
test_ids = set(item["id"] for item in test_data)

overlap_train_valid = train_ids & valid_ids
overlap_train_test = train_ids & test_ids
overlap_valid_test = valid_ids & test_ids

print("\nID 重疊檢查：")
print("train ∩ valid:", len(overlap_train_valid))
print("train ∩ test:", len(overlap_train_test))
print("valid ∩ test:", len(overlap_valid_test))

if overlap_train_valid or overlap_train_test or overlap_valid_test:
    raise ValueError("❌ train / valid / test 有 id 重疊，請檢查切分")

print("✅ train / valid / test 沒有資料重疊")

test_misleading_count = sum(
    1 for item in test_data
    if item["evidence_quality"] == "Misleading"
)

train_misleading_count = sum(
    1 for item in train_data
    if item["evidence_quality"] == "Misleading"
)

valid_misleading_count = sum(
    1 for item in valid_data
    if item["evidence_quality"] == "Misleading"
)

print("\nMisleading 分布：")
print("train_data:", train_misleading_count)
print("valid_data:", valid_misleading_count)
print("test_data:", test_misleading_count)

if test_misleading_count != len(misleading_data):
    raise ValueError("❌ Misleading 沒有全部放進 test")

print("✅ Misleading 已固定放進 test")

split_dist_df = pd.concat([
    show_split_distribution("train", train_data),
    show_split_distribution("valid", valid_data),
    show_split_distribution("test", test_data)
], ignore_index=True)

print("\nTrain / Valid / Test 標籤分布：")
display(split_dist_df)

split_save_content = {
    "split_seed": SPLIT_SEED,
    "test_size": TEST_SIZE,
    "valid_ratio_in_develop": VALID_RATIO_IN_DEVELOP,
    "train_data": train_data,
    "valid_data": valid_data,
    "test_data": test_data
}

with open(SPLIT_DATA_PATH, "w", encoding="utf-8") as f:
    json.dump(split_save_content, f, ensure_ascii=False, indent=2)

print("\n✅ 切分結果已儲存到：")
print(SPLIT_DATA_PATH)

print("\n✅ Cell 6 完成：已完成 train / valid / test 切分，且 Misleading 固定在 test")
````

### 讀回切分結果

**執行類型：必跑。**

重新讀取已儲存的 split 檔案，確認筆數、id 是否重疊，以及標籤分布是否合理。


````python
print("SPLIT_DATA_PATH:", SPLIT_DATA_PATH)

if not os.path.exists(SPLIT_DATA_PATH):
    raise FileNotFoundError(
        f"❌ 找不到 split 檔案：{SPLIT_DATA_PATH}\n"
        "請先執行 Cell 6 完成 train / valid / test 切分。"
    )

print("✅ 找到 split 檔案")

with open(SPLIT_DATA_PATH, "r", encoding="utf-8") as f:
    split_content = json.load(f)

train_data = split_content["train_data"]
valid_data = split_content["valid_data"]
test_data = split_content["test_data"]

print("\n✅ split 檔案讀取完成")
print("split_seed:", split_content.get("split_seed"))
print("test_size:", split_content.get("test_size"))
print("valid_ratio_in_develop:", split_content.get("valid_ratio_in_develop"))

print("\n資料筆數：")
print("train_data:", len(train_data))
print("valid_data:", len(valid_data))
print("test_data:", len(test_data))
print("total:", len(train_data) + len(valid_data) + len(test_data))

if len(train_data) != 700:
    raise ValueError(f"❌ train_data 筆數應為 700，目前是 {len(train_data)}")

if len(valid_data) != 100:
    raise ValueError(f"❌ valid_data 筆數應為 100，目前是 {len(valid_data)}")

if len(test_data) != 200:
    raise ValueError(f"❌ test_data 筆數應為 200，目前是 {len(test_data)}")

print("✅ train / valid / test 筆數正確")

train_ids = set(item["id"] for item in train_data)
valid_ids = set(item["id"] for item in valid_data)
test_ids = set(item["id"] for item in test_data)

print("\nID 重疊檢查：")
print("train ∩ valid:", len(train_ids & valid_ids))
print("train ∩ test:", len(train_ids & test_ids))
print("valid ∩ test:", len(valid_ids & test_ids))

if (train_ids & valid_ids) or (train_ids & test_ids) or (valid_ids & test_ids):
    raise ValueError("❌ train / valid / test 有 id 重疊")

print("✅ train / valid / test 沒有 id 重疊")

train_misleading = count_label(train_data, "evidence_quality", "Misleading")
valid_misleading = count_label(valid_data, "evidence_quality", "Misleading")
test_misleading = count_label(test_data, "evidence_quality", "Misleading")

print("\nMisleading 分布：")
print("train_data:", train_misleading)
print("valid_data:", valid_misleading)
print("test_data:", test_misleading)

if train_misleading != 0 or valid_misleading != 0 or test_misleading != 1:
    raise ValueError("❌ Misleading 分布不符合預期，應該只有 test_data 有 1 筆")

print("✅ Misleading 只在 test_data")

summary_rows = []

for split_name, data in [
    ("train", train_data),
    ("valid", valid_data),
    ("test", test_data)
]:
    counts = Counter(item["evidence_quality"] for item in data)

    for label in LABEL_LISTS["evidence_quality"]:
        summary_rows.append({
            "split": split_name,
            "label": label,
            "count": counts.get(label, 0),
            "ratio": round(counts.get(label, 0) / len(data), 4)
        })

summary_df = pd.DataFrame(summary_rows)

print("\nevidence_quality 分布：")
display(summary_df)

test_misleading_items = [
    item for item in test_data
    if item["evidence_quality"] == "Misleading"
]

print("\ntest_data 裡的 Misleading：")
for item in test_misleading_items:
    print("=" * 80)
    print("id:", item["id"])
    print("data:")
    print(item["data"])

print("\n✅ Cell 7 完成：split 檔案已讀回，資料切分確認正確")
````

### 回譯資料增強

**執行類型：重建增強資料時才跑。**

只對訓練資料做中文到英文再回中文的資料增強，並把結果存檔方便續跑。


````python
ZH_EN_MODEL_NAME = "Helsinki-NLP/opus-mt-zh-en"
EN_ZH_MODEL_NAME = "Helsinki-NLP/opus-mt-en-zh"

print("中文轉英文模型:", ZH_EN_MODEL_NAME)
print("英文轉中文模型:", EN_ZH_MODEL_NAME)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("使用裝置:", device)

print("\n📥 載入中文 → 英文模型...")
zh_en_tokenizer = MarianTokenizer.from_pretrained(ZH_EN_MODEL_NAME)
zh_en_model = MarianMTModel.from_pretrained(ZH_EN_MODEL_NAME).to(device)

print("📥 載入英文 → 中文模型...")
en_zh_tokenizer = MarianTokenizer.from_pretrained(EN_ZH_MODEL_NAME)
en_zh_model = MarianMTModel.from_pretrained(EN_ZH_MODEL_NAME).to(device)

zh_en_model.eval()
en_zh_model.eval()

print("✅ 回譯模型載入完成")

TEMP_AUGMENTED_PATH = AUGMENTED_DATA_PATH.replace(".json", "_temp.json")

print("AUGMENTED_DATA_PATH:", AUGMENTED_DATA_PATH)
print("TEMP_AUGMENTED_PATH:", TEMP_AUGMENTED_PATH)

if os.path.exists(TEMP_AUGMENTED_PATH):
    print("\n✅ 找到暫存回譯檔，從暫存檔繼續：")
    print(TEMP_AUGMENTED_PATH)

    with open(TEMP_AUGMENTED_PATH, "r", encoding="utf-8") as f:
        train_data_augmented = json.load(f)

else:
    print("\n📌 沒有暫存檔，從 train_data 開始建立回譯資料")

    train_data_augmented = copy.deepcopy(train_data)

    for item in train_data_augmented:
        item["data_aug"] = None

print("目前 train_data_augmented 筆數:", len(train_data_augmented))

unfinished_indices = [
    idx for idx, item in enumerate(train_data_augmented)
    if item.get("data_aug") is None or str(item.get("data_aug")).strip() == ""
]

print("尚未完成回譯筆數:", len(unfinished_indices))
print("已完成回譯筆數:", len(train_data_augmented) - len(unfinished_indices))

BATCH_SIZE_TRANSLATE = 4
SAVE_EVERY_BATCH = 10

print("\n開始回譯 train_data")
print("翻譯 batch size:", BATCH_SIZE_TRANSLATE)
print("每幾個 batch 存一次:", SAVE_EVERY_BATCH)

for batch_start in tqdm(
    range(0, len(unfinished_indices), BATCH_SIZE_TRANSLATE),
    desc="Back Translation"
):
    batch_indices = unfinished_indices[
        batch_start : batch_start + BATCH_SIZE_TRANSLATE
    ]

    batch_texts = [
        train_data_augmented[idx]["data"]
        for idx in batch_indices
    ]

    try:
        batch_aug_texts = back_translate_batch(batch_texts)

        for idx, aug_text in zip(batch_indices, batch_aug_texts):
            train_data_augmented[idx]["data_aug"] = aug_text

    except Exception as e:
        print("\n⚠️ 回譯時發生錯誤")
        print("batch_start:", batch_start)
        print("錯誤訊息:", e)

        for idx in batch_indices:
            train_data_augmented[idx]["data_aug"] = train_data_augmented[idx]["data"]

    batch_num = batch_start // BATCH_SIZE_TRANSLATE

    if batch_num % SAVE_EVERY_BATCH == 0:
        with open(TEMP_AUGMENTED_PATH, "w", encoding="utf-8") as f:
            json.dump(train_data_augmented, f, ensure_ascii=False, indent=2)

with open(AUGMENTED_DATA_PATH, "w", encoding="utf-8") as f:
    json.dump(train_data_augmented, f, ensure_ascii=False, indent=2)

print("\n✅ 回譯完成，正式檔案已儲存：")
print(AUGMENTED_DATA_PATH)

missing_aug_count = sum(
    1 for item in train_data_augmented
    if item.get("data_aug") is None or str(item.get("data_aug")).strip() == ""
)

print("\n回譯後檢查：")
print("train_data_augmented 筆數:", len(train_data_augmented))
print("data_aug 缺失筆數:", missing_aug_count)

if missing_aug_count > 0:
    raise ValueError("❌ 還有 data_aug 缺失，請檢查回譯流程")
else:
    print("✅ 所有 train_data 都有 data_aug")

print("\n原文 vs 回譯範例：")
for i in range(3):
    print("=" * 80)
    print("id:", train_data_augmented[i]["id"])
    print("\n原文 data:")
    print(train_data_augmented[i]["data"])
    print("\n回譯 data_aug:")
    print(train_data_augmented[i]["data_aug"])

print("\n✅ Cell 8 完成：train_data 回譯增強完成，valid/test 沒有被動到")
````

### 檢查回譯資料

讀取回譯結果，確認欄位、id 對應與增強後文字是否可用。


````python
print("AUGMENTED_DATA_PATH:", AUGMENTED_DATA_PATH)

if not os.path.exists(AUGMENTED_DATA_PATH):
    raise FileNotFoundError(
        f"❌ 找不到回譯檔案：{AUGMENTED_DATA_PATH}\n"
        "請先執行 Cell 8 完成回譯資料增強。"
    )

print("✅ 找到回譯檔案")

with open(AUGMENTED_DATA_PATH, "r", encoding="utf-8") as f:
    train_data_augmented = json.load(f)

print("\n✅ 回譯資料讀取完成")
print("train_data_augmented 筆數:", len(train_data_augmented))

if len(train_data_augmented) != len(train_data):
    raise ValueError(
        f"❌ 回譯資料筆數不等於 train_data\n"
        f"train_data: {len(train_data)}\n"
        f"train_data_augmented: {len(train_data_augmented)}"
    )

print("✅ 回譯資料筆數正確")

REQUIRED_AUG_FIELDS = [
    "id",
    "data",
    "data_aug",
    "promise_status",
    "verification_timeline",
    "evidence_status",
    "evidence_quality"
]

bad_records = []

for idx, item in enumerate(train_data_augmented):
    missing_fields = [
        field for field in REQUIRED_AUG_FIELDS
        if field not in item
    ]

    if len(missing_fields) > 0:
        bad_records.append({
            "index": idx,
            "id": item.get("id", None),
            "problem": "missing_fields",
            "detail": missing_fields
        })
        continue

    if not isinstance(item["data"], str) or item["data"].strip() == "":
        bad_records.append({
            "index": idx,
            "id": item.get("id", None),
            "problem": "bad_data",
            "detail": item.get("data", None)
        })

    if not isinstance(item["data_aug"], str) or item["data_aug"].strip() == "":
        bad_records.append({
            "index": idx,
            "id": item.get("id", None),
            "problem": "bad_data_aug",
            "detail": item.get("data_aug", None)
        })

    for task in TASK_FIELDS:
        if item.get(task) not in LABEL_LISTS[task]:
            bad_records.append({
                "index": idx,
                "id": item.get("id", None),
                "problem": f"invalid_label_{task}",
                "detail": item.get(task)
            })

if len(bad_records) > 0:
    print("\n❌ 回譯資料格式有問題：")
    display(pd.DataFrame(bad_records).head(50))
    raise ValueError("回譯資料格式檢查失敗")
else:
    print("✅ 每筆回譯資料都有 data / data_aug / 四個任務標籤")

train_ids = set(item["id"] for item in train_data)
aug_ids = set(item["id"] for item in train_data_augmented)

print("\nid 一致性檢查：")
print("train_data ids:", len(train_ids))
print("train_data_augmented ids:", len(aug_ids))
print("差異數:", len(train_ids ^ aug_ids))

if train_ids != aug_ids:
    raise ValueError("❌ train_data 和 train_data_augmented 的 id 不一致")

print("✅ train_data 和 train_data_augmented 的 id 完全一致")

valid_has_aug = sum(
    1 for item in valid_data
    if "data_aug" in item
)

test_has_aug = sum(
    1 for item in test_data
    if "data_aug" in item
)

print("\nvalid/test data_aug 檢查：")
print("valid_data 有 data_aug 的筆數:", valid_has_aug)
print("test_data 有 data_aug 的筆數:", test_has_aug)

if valid_has_aug > 0 or test_has_aug > 0:
    print("⚠️ valid/test 有 data_aug 欄位，這不一定錯，但訓練時不會使用")
else:
    print("✅ valid/test 沒有被回譯污染")

print("\n回譯前後範例：")

for i in range(3):
    item = train_data_augmented[i]

    print("=" * 80)
    print("id:", item["id"])
    print("promise_status:", item["promise_status"])
    print("evidence_status:", item["evidence_status"])
    print("evidence_quality:", item["evidence_quality"])

    print("\n原文 data:")
    print(item["data"])

    print("\n回譯 data_aug:")
    print(item["data_aug"])

train_data_for_model = train_data_augmented

print("\n✅ train_data_for_model 已建立")
print("train_data_for_model 筆數:", len(train_data_for_model))

print("\n✅ Cell 9 完成：回譯資料讀取與格式檢查完成")
````

### 建立訓練資料版本

**執行類型：必跑。**

建立 original、augmented-only 與 mixed 三種訓練版本，並選定後續模型使用的版本。


````python
train_original = []

for item in train_data_for_model:
    new_item = copy.deepcopy(item)
    new_item["input_text"] = new_item["data"]
    new_item["source_type"] = "original"
    train_original.append(new_item)

print("train_original 筆數:", len(train_original))

train_aug_only = []

for item in train_data_for_model:
    new_item = copy.deepcopy(item)
    new_item["input_text"] = new_item["data_aug"]
    new_item["source_type"] = "aug_only"
    train_aug_only.append(new_item)

print("train_aug_only 筆數:", len(train_aug_only))

train_full_mix = []

for item in train_data_for_model:
    original_item = copy.deepcopy(item)
    original_item["input_text"] = original_item["data"]
    original_item["source_type"] = "original"
    train_full_mix.append(original_item)

    aug_item = copy.deepcopy(item)
    aug_item["input_text"] = aug_item["data_aug"]
    aug_item["source_type"] = "augmented"
    train_full_mix.append(aug_item)

print("train_full_mix 筆數:", len(train_full_mix))

valid_for_model = []

for item in valid_data:
    new_item = copy.deepcopy(item)
    new_item["input_text"] = new_item["data"]
    new_item["source_type"] = "valid_original"
    valid_for_model.append(new_item)

test_for_model = []

for item in test_data:
    new_item = copy.deepcopy(item)
    new_item["input_text"] = new_item["data"]
    new_item["source_type"] = "test_original"
    test_for_model.append(new_item)

print("\nvalid_for_model 筆數:", len(valid_for_model))
print("test_for_model 筆數:", len(test_for_model))

TRAIN_VERSION = "full_mix"

if TRAIN_VERSION == "original":
    final_train_for_model = train_original

elif TRAIN_VERSION == "aug_only":
    final_train_for_model = train_aug_only

elif TRAIN_VERSION == "full_mix":
    final_train_for_model = train_full_mix

else:
    raise ValueError("❌ TRAIN_VERSION 必須是 original / aug_only / full_mix")

print("\n✅ 目前選擇的 TRAIN_VERSION:", TRAIN_VERSION)
print("final_train_for_model 筆數:", len(final_train_for_model))

bad_input_records = []

for idx, item in enumerate(final_train_for_model):
    if "input_text" not in item:
        bad_input_records.append({
            "index": idx,
            "id": item.get("id", None),
            "problem": "missing_input_text"
        })

    elif not isinstance(item["input_text"], str) or item["input_text"].strip() == "":
        bad_input_records.append({
            "index": idx,
            "id": item.get("id", None),
            "problem": "bad_input_text",
            "input_text": item.get("input_text", None)
        })

if len(bad_input_records) > 0:
    print("\n❌ input_text 有問題：")
    display(pd.DataFrame(bad_input_records).head(20))
    raise ValueError("final_train_for_model 有 input_text 問題")
else:
    print("✅ final_train_for_model 的 input_text 都正常")

dist_compare_df = pd.concat([
    build_distribution_df(train_original, "train_original"),
    build_distribution_df(train_aug_only, "train_aug_only"),
    build_distribution_df(train_full_mix, "train_full_mix"),
    build_distribution_df(valid_for_model, "valid"),
    build_distribution_df(test_for_model, "test")
], ignore_index=True)

print("\n各版本資料標籤分布：")
display(dist_compare_df)

print("\nfinal_train_for_model 範例：")

for i in range(3):
    item = final_train_for_model[i]

    print("=" * 80)
    print("id:", item["id"])
    print("source_type:", item["source_type"])
    print("promise_status:", item["promise_status"])
    print("evidence_status:", item["evidence_status"])
    print("evidence_quality:", item["evidence_quality"])
    print("\ninput_text:")
    print(item["input_text"][:500])

print("\n✅ Cell 10 完成：三組訓練資料版本已建立，最終訓練版本為", TRAIN_VERSION)
````

### 模型結構測試

**執行類型：檢查用；確認模型 shape 後可跳過。**

建立暫時 RoBERTa 模型跑過一個 batch，確認多任務輸出 shape 正確。


````python
test_model = MultiTaskRoBERTa(
    model_name=MODEL_NAME,
    task_fields=ROBERTA_TASK_FIELDS,
    label_lists=LABEL_LISTS,
    dropout_rate=DROPOUT_RATE
).to(device)

print("\n✅ 測試用 RoBERTa 模型已建立")

batch = next(iter(train_loader))

input_ids = batch["input_ids"].to(device)
attention_mask = batch["attention_mask"].to(device)

test_model.eval()

with torch.no_grad():
    logits = test_model(
        input_ids=input_ids,
        attention_mask=attention_mask
    )

print("\n模型輸出 logits 檢查：")

for task in ROBERTA_TASK_FIELDS:
    print(task, "logits shape:", logits[task].shape)

    expected_num_labels = len(LABEL_LISTS[task])

    if logits[task].shape[0] != BATCH_SIZE:
        raise ValueError(f"❌ {task} batch size 不正確")

    if logits[task].shape[1] != expected_num_labels:
        raise ValueError(f"❌ {task} label 數量不正確")

print("✅ logits shape 正確")

del test_model

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    print("✅ GPU cache 已清除")

print("\n✅ Cell 13 完成：RoBERTa 多任務模型建立成功")
````

### Loss 與評估流程測試

### 建立 Retrieval Pool

**執行類型：必跑。**

將 train 與 valid 整理成 RAG 可檢索的範例池，供後續 prompt 使用。


````python
retrieval_pool = []

for item in train_data:
    new_item = copy.deepcopy(item)
    new_item["source_split"] = "train"
    new_item["retrieval_text"] = new_item["data"]
    retrieval_pool.append(new_item)

for item in valid_data:
    new_item = copy.deepcopy(item)
    new_item["source_split"] = "valid"
    new_item["retrieval_text"] = new_item["data"]
    retrieval_pool.append(new_item)

print("retrieval_pool 筆數:", len(retrieval_pool))
print("train_data 筆數:", len(train_data))
print("valid_data 筆數:", len(valid_data))
print("test_data 筆數:", len(test_data))

if len(retrieval_pool) != len(train_data) + len(valid_data):
    raise ValueError("❌ retrieval_pool 筆數不正確")

print("✅ retrieval_pool = train + valid，沒有包含 test")

retrieval_ids = set(item["id"] for item in retrieval_pool)
test_ids = set(item["id"] for item in test_data)

overlap_with_test = retrieval_ids & test_ids

print("\n與 test_data 的 id 重疊數:", len(overlap_with_test))

if len(overlap_with_test) > 0:
    print("❌ 發現 retrieval_pool 和 test_data 有重疊 id：")
    print(list(overlap_with_test)[:20])
    raise ValueError("retrieval_pool 不能包含 test_data")
else:
    print("✅ retrieval_pool 沒有包含任何 test_data")

bad_text_records = []

for idx, item in enumerate(retrieval_pool):
    if "retrieval_text" not in item:
        bad_text_records.append({
            "index": idx,
            "id": item.get("id", None),
            "problem": "missing_retrieval_text"
        })

    elif not isinstance(item["retrieval_text"], str) or item["retrieval_text"].strip() == "":
        bad_text_records.append({
            "index": idx,
            "id": item.get("id", None),
            "problem": "bad_retrieval_text",
            "retrieval_text": item.get("retrieval_text", None)
        })

if len(bad_text_records) > 0:
    print("\n❌ retrieval_text 有問題：")
    display(pd.DataFrame(bad_text_records).head(20))
    raise ValueError("retrieval_pool 有 retrieval_text 問題")
else:
    print("✅ retrieval_pool 每筆都有有效 retrieval_text")

distribution_rows = []

for task in TASK_FIELDS:
    counts = Counter(item[task] for item in retrieval_pool)
    total = len(retrieval_pool)

    for label in LABEL_LISTS[task]:
        distribution_rows.append({
            "task": task,
            "label": label,
            "count": counts.get(label, 0),
            "ratio": round(counts.get(label, 0) / total, 4)
        })

retrieval_dist_df = pd.DataFrame(distribution_rows)

print("\nretrieval_pool 標籤分布：")
display(retrieval_dist_df)

retrieval_misleading_count = sum(
    1 for item in retrieval_pool
    if item["evidence_quality"] == "Misleading"
)

test_misleading_count = sum(
    1 for item in test_data
    if item["evidence_quality"] == "Misleading"
)

print("\nMisleading 檢查：")
print("retrieval_pool Misleading 筆數:", retrieval_misleading_count)
print("test_data Misleading 筆數:", test_misleading_count)

if retrieval_misleading_count > 0:
    print("⚠️ retrieval_pool 有 Misleading，請確認是否符合你的實驗設定")
else:
    print("✅ retrieval_pool 沒有 Misleading，避免 test leakage")

with open(RETRIEVAL_POOL_PATH, "w", encoding="utf-8") as f:
    json.dump(retrieval_pool, f, ensure_ascii=False, indent=2)

print("\n✅ retrieval_pool 已儲存：")
print(RETRIEVAL_POOL_PATH)

print("\nretrieval_pool 前 3 筆預覽：")

for i in range(3):
    item = retrieval_pool[i]

    print("=" * 80)
    print("id:", item["id"])
    print("source_split:", item["source_split"])
    print("promise_status:", item["promise_status"])
    print("verification_timeline:", item["verification_timeline"])
    print("evidence_status:", item["evidence_status"])
    print("evidence_quality:", item["evidence_quality"])
    print("\nretrieval_text:")
    print(item["retrieval_text"][:500])

print("\n✅ Cell 20 完成：RAG retrieval_pool 已建立完成")
````

### 測試 Retrieval Top-K

**執行類型：建議跑；確認 RAG 的 Top-K 設定。**

排除 self-match 後比較不同 Top-K retrieval baseline，選出後續 GPT + RAG 使用的設定。


````python
TOP_K_LIST = [3, 5, 6, 8, 10]

RAG_TASK_FIELDS = [
    "verification_timeline",
    "evidence_quality"
]

print("重新比較 TOP_K_LIST:", TOP_K_LIST)
print("這次會排除 valid 自己的 id，避免 self-match")

topk_no_self_results = []
topk_no_self_prediction_dict = {}

for top_k in TOP_K_LIST:
    print("\n" + "=" * 80)
    print(f"重新測試 top_k = {top_k}，排除 self-match")
    print("=" * 80)

    preds = retrieval_majority_baseline_no_self(
        eval_data=valid_data,
        top_k=top_k
    )

    topk_no_self_prediction_dict[top_k] = preds

    result_row = {
        "top_k": top_k
    }

    for task in RAG_TASK_FIELDS:
        y_true = [
            item[f"{task}_true"] for item in preds
        ]

        y_pred = [
            item[f"{task}_pred"] for item in preds
        ]

        labels = LABEL_LISTS[task]

        macro_f1 = f1_score(
            y_true,
            y_pred,
            labels=labels,
            average="macro",
            zero_division=0
        )

        micro_f1 = f1_score(
            y_true,
            y_pred,
            labels=labels,
            average="micro",
            zero_division=0
        )

        acc = accuracy_score(y_true, y_pred)

        result_row[f"{task}_macro_f1"] = macro_f1
        result_row[f"{task}_micro_f1"] = micro_f1
        result_row[f"{task}_accuracy"] = acc

        print("\n" + "-" * 60)
        print(f"{task} | top_k = {top_k}")
        print("-" * 60)
        print("Macro F1:", round(macro_f1, 4))
        print("Micro F1:", round(micro_f1, 4))
        print("Accuracy:", round(acc, 4))

    partial_score = (
        result_row["verification_timeline_macro_f1"] * OFFICIAL_WEIGHTS["verification_timeline"]
        + result_row["evidence_quality_macro_f1"] * OFFICIAL_WEIGHTS["evidence_quality"]
    )

    result_row["rag_partial_score"] = partial_score

    print("\nNo-self RAG Retrieval Baseline Partial Score:", round(partial_score, 6))

    topk_no_self_results.append(result_row)

topk_no_self_results_df = pd.DataFrame(topk_no_self_results)

print("\n排除 self-match 後的 Top-K 比較結果：")
display(topk_no_self_results_df)

best_topk_no_self_row = topk_no_self_results_df.sort_values(
    "rag_partial_score",
    ascending=False
).iloc[0]

BEST_TOP_K = int(best_topk_no_self_row["top_k"])

print("\n✅ 排除 self-match 後，表現最好的 top_k:")
print("BEST_TOP_K:", BEST_TOP_K)
print("rag_partial_score:", best_topk_no_self_row["rag_partial_score"])

TOPK_NO_SELF_RESULT_PATH = os.path.join(
    OUTPUT_DIR,
    "rag_topk_retrieval_baseline_no_self_results.csv"
)

topk_no_self_results_df.to_csv(
    TOPK_NO_SELF_RESULT_PATH,
    index=False,
    encoding="utf-8-sig"
)

print("\n✅ No-self Top-K 比較結果已儲存：")
print(TOPK_NO_SELF_RESULT_PATH)

TOPK_NO_SELF_DETAIL_PATH = os.path.join(
    OUTPUT_DIR,
    "rag_topk_retrieval_baseline_no_self_details.json"
)

topk_no_self_prediction_dict_serializable = {
    str(k): v for k, v in topk_no_self_prediction_dict.items()
}

with open(TOPK_NO_SELF_DETAIL_PATH, "w", encoding="utf-8") as f:
    json.dump(topk_no_self_prediction_dict_serializable, f, ensure_ascii=False, indent=2)

print("\n✅ No-self Top-K 詳細預測已儲存：")
print(TOPK_NO_SELF_DETAIL_PATH)

print("\n最佳 top_k 的前 3 筆 valid 預測範例：")

best_no_self_preds = topk_no_self_prediction_dict[BEST_TOP_K]

for item in best_no_self_preds[:3]:
    print("=" * 80)
    print("id:", item["id"])
    print("top_k:", item["top_k"])

    for task in RAG_TASK_FIELDS:
        print(f"\n{task}")
        print("true:", item[f"{task}_true"])
        print("pred:", item[f"{task}_pred"])
        print("retrieved_labels:", item[f"{task}_retrieved_labels"])

    print("\nretrieved_ids:", item["retrieved_ids"])
    print("similarities:", [round(s, 4) for s in item["retrieved_similarities"]])

print("\n✅ Cell 22.5 完成：已排除 self-match 並重新選出 BEST_TOP_K")
````

````python
GPT_RAG_PROGRESS_PATH = os.path.join(
    OUTPUT_DIR,
    "gpt_rag_task24_progress.json"
)

GPT_RAG_TASK24_PATH = os.path.join(
    OUTPUT_DIR,
    "gpt_rag_task24_predictions.json"
)

print("讀取 GPT 預測結果：")
print(GPT_RAG_PROGRESS_PATH)

if not os.path.exists(GPT_RAG_PROGRESS_PATH):
    raise FileNotFoundError(
        f"❌ 找不到 GPT 進度檔：{GPT_RAG_PROGRESS_PATH}"
    )

with open(GPT_RAG_PROGRESS_PATH, "r", encoding="utf-8") as f:
    gpt_rag_predictions = json.load(f)

print("目前 GPT 預測筆數:", len(gpt_rag_predictions))
print("test_data 筆數:", len(test_data))

error_records = [
    item for item in gpt_rag_predictions
    if "error" in item
]

success_records = [
    item for item in gpt_rag_predictions
    if "error" not in item
]

print("\n成功筆數:", len(success_records))
print("失敗筆數:", len(error_records))

pred_ids = [
    normalize_id(item["id"])
    for item in gpt_rag_predictions
]

test_ids = [
    normalize_id(item["id"])
    for item in test_data
]

duplicate_ids = [
    item_id for item_id, count in Counter(pred_ids).items()
    if count > 1
]

missing_ids = sorted(
    list(set(test_ids) - set(pred_ids))
)

extra_ids = sorted(
    list(set(pred_ids) - set(test_ids))
)

print("\nID 檢查：")
print("重複 id 數:", len(duplicate_ids))
print("漏掉 id 數:", len(missing_ids))
print("多出 id 數:", len(extra_ids))

if len(duplicate_ids) > 0:
    print("重複 id 前 20 個:", duplicate_ids[:20])

if len(missing_ids) > 0:
    print("漏掉 id 前 20 個:", missing_ids[:20])

if len(extra_ids) > 0:
    print("多出 id 前 20 個:", extra_ids[:20])

if len(error_records) > 0:
    print("\n前 10 筆失敗 id：")
    for item in error_records[:10]:
        print(item["id"], "=>", item.get("error"))

    error_counter = Counter(
        item.get("error", "unknown")
        for item in error_records
    )

    print("\n錯誤類型統計：")
    for err, count in error_counter.items():
        print("=" * 80)
        print("count:", count)
        print("error:", err[:500])

    error_df = pd.DataFrame(error_records)

    display_cols = [
        "id",
        "verification_timeline_true",
        "verification_timeline_pred",
        "evidence_quality_true",
        "evidence_quality_pred",
        "error"
    ]

    existing_display_cols = [
        col for col in display_cols
        if col in error_df.columns
    ]

    print("\n失敗資料表格前 20 筆：")
    display(error_df[existing_display_cols].head(20))

else:
    print("\n✅ 沒有失敗資料，可以進入 Cell 26 算正式分數")

FAILED_GPT_RECORDS_PATH = os.path.join(
    OUTPUT_DIR,
    "gpt_rag_failed_records.json"
)

with open(FAILED_GPT_RECORDS_PATH, "w", encoding="utf-8") as f:
    json.dump(error_records, f, ensure_ascii=False, indent=2)

print("\n✅ 失敗資料清單已儲存：")
print(FAILED_GPT_RECORDS_PATH)

print("\n✅ Cell 25.5 完成：GPT 預測成功/失敗狀態檢查完成")
````

### GPT 任務分數

**執行類型：必跑；評估 Task 2/4。**

計算 GPT 負責的 verification timeline 與 evidence quality 分數。


````python
GPT_RAG_TASK24_PATH = os.path.join(
    OUTPUT_DIR,
    "gpt_rag_task24_predictions.json"
)

GPT_RAG_PROGRESS_PATH = os.path.join(
    OUTPUT_DIR,
    "gpt_rag_task24_progress.json"
)

if os.path.exists(GPT_RAG_TASK24_PATH):
    load_path = GPT_RAG_TASK24_PATH
else:
    load_path = GPT_RAG_PROGRESS_PATH

print("讀取 GPT + RAG 預測結果：")
print(load_path)

with open(load_path, "r", encoding="utf-8") as f:
    gpt_rag_predictions = json.load(f)

print("GPT + RAG 預測筆數:", len(gpt_rag_predictions))
print("test_data 筆數:", len(test_data))

if len(gpt_rag_predictions) != len(test_data):
    print("⚠️ 預測筆數與 test_data 不一致，請先檢查")
else:
    print("✅ 預測筆數正確")

error_records = [
    item for item in gpt_rag_predictions
    if "error" in item
]

print("error records 數量:", len(error_records))

if len(error_records) > 0:
    print("⚠️ 仍有 error records，這會影響正式分數")
    print("前 5 筆 error:")
    for item in error_records[:5]:
        print(item["id"], item.get("error"))
else:
    print("✅ 沒有 error records")

gpt_eval_rows = []

for item in gpt_rag_predictions:
    row = {
        "id": normalize_id(item["id"]),
        "data": item.get("data", ""),

        "verification_timeline_true": item["verification_timeline_true"],
        "verification_timeline_pred": item["verification_timeline_pred"],

        "evidence_quality_true": item["evidence_quality_true"],
        "evidence_quality_pred": item["evidence_quality_pred"],

        "promise_status_roberta_pred": item.get("promise_status_roberta_pred"),
        "evidence_status_roberta_pred": item.get("evidence_status_roberta_pred"),

        "top_k": item.get("top_k"),
        "gpt_model": item.get("gpt_model", GPT_MODEL)
    }

    gpt_eval_rows.append(row)

gpt_eval_df = pd.DataFrame(gpt_eval_rows)

print("\nGPT eval dataframe shape:", gpt_eval_df.shape)
display(gpt_eval_df.head())

timeline_y_true = gpt_eval_df["verification_timeline_true"].tolist()
timeline_y_pred = gpt_eval_df["verification_timeline_pred"].tolist()

timeline_macro_f1 = f1_score(
    timeline_y_true,
    timeline_y_pred,
    labels=VALID_VERIFICATION_TIMELINE,
    average="macro",
    zero_division=0
)

timeline_micro_f1 = f1_score(
    timeline_y_true,
    timeline_y_pred,
    labels=VALID_VERIFICATION_TIMELINE,
    average="micro",
    zero_division=0
)

timeline_acc = accuracy_score(
    timeline_y_true,
    timeline_y_pred
)

print("\n" + "=" * 80)
print("Task 2: verification_timeline")
print("=" * 80)

print("Macro F1:", round(timeline_macro_f1, 4))
print("Micro F1:", round(timeline_micro_f1, 4))
print("Accuracy:", round(timeline_acc, 4))

print("\nClassification Report:")
print(
    classification_report(
        timeline_y_true,
        timeline_y_pred,
        labels=VALID_VERIFICATION_TIMELINE,
        zero_division=0
    )
)

quality_y_true = gpt_eval_df["evidence_quality_true"].tolist()
quality_y_pred = gpt_eval_df["evidence_quality_pred"].tolist()

quality_macro_f1 = f1_score(
    quality_y_true,
    quality_y_pred,
    labels=VALID_EVIDENCE_QUALITY,
    average="macro",
    zero_division=0
)

quality_micro_f1 = f1_score(
    quality_y_true,
    quality_y_pred,
    labels=VALID_EVIDENCE_QUALITY,
    average="micro",
    zero_division=0
)

quality_acc = accuracy_score(
    quality_y_true,
    quality_y_pred
)

print("\n" + "=" * 80)
print("Task 4: evidence_quality")
print("=" * 80)

print("Macro F1:", round(quality_macro_f1, 4))
print("Micro F1:", round(quality_micro_f1, 4))
print("Accuracy:", round(quality_acc, 4))

print("\nClassification Report:")
print(
    classification_report(
        quality_y_true,
        quality_y_pred,
        labels=VALID_EVIDENCE_QUALITY,
        zero_division=0
    )
)

gpt_rag_partial_score = (
    timeline_macro_f1 * OFFICIAL_WEIGHTS["verification_timeline"]
    + quality_macro_f1 * OFFICIAL_WEIGHTS["evidence_quality"]
)

print("\n" + "=" * 80)
print("GPT + RAG Task 2 + Task 4 Partial Score")
print("=" * 80)

print("verification_timeline weight:", OFFICIAL_WEIGHTS["verification_timeline"])
print("evidence_quality weight:", OFFICIAL_WEIGHTS["evidence_quality"])
print("GPT + RAG partial score:", round(gpt_rag_partial_score, 6))

gpt_rag_eval_result = {
    "gpt_model": GPT_MODEL,
    "top_k": BEST_TOP_K,

    "verification_timeline": {
        "macro_f1": timeline_macro_f1,
        "micro_f1": timeline_micro_f1,
        "accuracy": timeline_acc,
        "weight": OFFICIAL_WEIGHTS["verification_timeline"]
    },

    "evidence_quality": {
        "macro_f1": quality_macro_f1,
        "micro_f1": quality_micro_f1,
        "accuracy": quality_acc,
        "weight": OFFICIAL_WEIGHTS["evidence_quality"]
    },

    "gpt_rag_partial_score": gpt_rag_partial_score,
    "num_predictions": len(gpt_rag_predictions),
    "num_errors": len(error_records)
}

GPT_RAG_EVAL_RESULT_PATH = os.path.join(
    OUTPUT_DIR,
    "gpt_rag_task24_eval_results.json"
)

with open(GPT_RAG_EVAL_RESULT_PATH, "w", encoding="utf-8") as f:
    json.dump(gpt_rag_eval_result, f, ensure_ascii=False, indent=2)

print("\n✅ GPT + RAG 評估結果已儲存：")
print(GPT_RAG_EVAL_RESULT_PATH)

GPT_RAG_EVAL_CSV_PATH = os.path.join(
    OUTPUT_DIR,
    "gpt_rag_task24_eval_predictions.csv"
)

gpt_eval_df.to_csv(
    GPT_RAG_EVAL_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

print("\n✅ GPT + RAG 逐筆預測 CSV 已儲存：")
print(GPT_RAG_EVAL_CSV_PATH)

print("\n✅ Cell 26 完成：GPT + RAG Task 2 / Task 4 分數計算完成")
````

## 5. 最終評估流程

整合 RoBERTa Task 1/3 與 GPT/RAG v4 Task 2/4，計算 hybrid final score，並套用任務依賴規則。

### Hybrid 最終分數

整合四任務預測結果並計算官方總分。


````python
ROBERTA_ENSEMBLE_OUTPUT_PATH = os.path.join(
    OUTPUT_DIR,
    "roberta_task13_ensemble_predictions.json"
)

print("讀取 RoBERTa ensemble 結果：")
print(ROBERTA_ENSEMBLE_OUTPUT_PATH)

if not os.path.exists(ROBERTA_ENSEMBLE_OUTPUT_PATH):
    raise FileNotFoundError(
        f"❌ 找不到 RoBERTa ensemble 結果：{ROBERTA_ENSEMBLE_OUTPUT_PATH}"
    )

with open(ROBERTA_ENSEMBLE_OUTPUT_PATH, "r", encoding="utf-8") as f:
    roberta_output_content = json.load(f)

roberta_test_predictions = roberta_output_content["test_predictions"]

print("RoBERTa test predictions 筆數:", len(roberta_test_predictions))

GPT_RAG_TASK24_PATH = os.path.join(
    OUTPUT_DIR,
    "gpt_rag_task24_predictions.json"
)

print("\n讀取 GPT + RAG 結果：")
print(GPT_RAG_TASK24_PATH)

if not os.path.exists(GPT_RAG_TASK24_PATH):
    raise FileNotFoundError(
        f"❌ 找不到 GPT + RAG 預測結果：{GPT_RAG_TASK24_PATH}"
    )

with open(GPT_RAG_TASK24_PATH, "r", encoding="utf-8") as f:
    gpt_rag_predictions = json.load(f)

print("GPT + RAG predictions 筆數:", len(gpt_rag_predictions))

roberta_by_id = {
    normalize_id(item["id"]): item
    for item in roberta_test_predictions
}

gpt_by_id = {
    normalize_id(item["id"]): item
    for item in gpt_rag_predictions
}

test_by_id = {
    normalize_id(item["id"]): item
    for item in test_data
}

print("\nID 數量檢查：")
print("test_by_id:", len(test_by_id))
print("roberta_by_id:", len(roberta_by_id))
print("gpt_by_id:", len(gpt_by_id))

missing_roberta_ids = set(test_by_id.keys()) - set(roberta_by_id.keys())
missing_gpt_ids = set(test_by_id.keys()) - set(gpt_by_id.keys())

print("缺 RoBERTa 預測 id 數:", len(missing_roberta_ids))
print("缺 GPT 預測 id 數:", len(missing_gpt_ids))

if len(missing_roberta_ids) > 0:
    print("缺 RoBERTa 前 20 個:", list(missing_roberta_ids)[:20])
    raise ValueError("❌ 有 test_data 找不到 RoBERTa 預測")

if len(missing_gpt_ids) > 0:
    print("缺 GPT 前 20 個:", list(missing_gpt_ids)[:20])
    raise ValueError("❌ 有 test_data 找不到 GPT 預測")

print("✅ 四任務 id 對應成功")

hybrid_final_predictions = []

for item in test_data:
    item_id = normalize_id(item["id"])

    roberta_item = roberta_by_id[item_id]
    gpt_item = gpt_by_id[item_id]

    final_item = {
        "id": item_id,
        "data": item["data"],

        # True labels
        "promise_status_true": item["promise_status"],
        "verification_timeline_true": item["verification_timeline"],
        "evidence_status_true": item["evidence_status"],
        "evidence_quality_true": item["evidence_quality"],

        "promise_status_pred": roberta_item["promise_status_pred"],
        "verification_timeline_pred": gpt_item["verification_timeline_pred"],
        "evidence_status_pred": roberta_item["evidence_status_pred"],
        "evidence_quality_pred": gpt_item["evidence_quality_pred"],

        "promise_status_method": "RoBERTa ensemble",
        "verification_timeline_method": f"GPT + RAG ({gpt_item.get('gpt_model', GPT_MODEL)})",
        "evidence_status_method": "RoBERTa ensemble",
        "evidence_quality_method": f"GPT + RAG ({gpt_item.get('gpt_model', GPT_MODEL)})",

        "top_k": gpt_item.get("top_k", BEST_TOP_K),
        "gpt_model": gpt_item.get("gpt_model", GPT_MODEL)
    }

    final_item = apply_task_dependency_rules(final_item)
    hybrid_final_predictions.append(final_item)

print("\nHybrid final predictions 筆數:", len(hybrid_final_predictions))

if len(hybrid_final_predictions) != len(test_data):
    raise ValueError("❌ 整合後筆數不等於 test_data")

print("✅ 四任務預測整合完成，並已套用任務依賴規則")

FINAL_TASK_FIELDS = [
    "promise_status",
    "verification_timeline",
    "evidence_status",
    "evidence_quality"
]

final_eval_results = {}
final_total_score = 0.0

for task in FINAL_TASK_FIELDS:
    y_true = [
        item[f"{task}_true"]
        for item in hybrid_final_predictions
    ]

    y_pred = [
        item[f"{task}_pred"]
        for item in hybrid_final_predictions
    ]

    labels = LABEL_LISTS[task]

    macro_f1 = f1_score(
        y_true,
        y_pred,
        labels=labels,
        average="macro",
        zero_division=0
    )

    micro_f1 = f1_score(
        y_true,
        y_pred,
        labels=labels,
        average="micro",
        zero_division=0
    )

    acc = accuracy_score(y_true, y_pred)

    weighted_score = macro_f1 * OFFICIAL_WEIGHTS[task]
    final_total_score += weighted_score

    final_eval_results[task] = {
        "macro_f1": macro_f1,
        "micro_f1": micro_f1,
        "accuracy": acc,
        "weight": OFFICIAL_WEIGHTS[task],
        "weighted_score": weighted_score
    }

print("\n" + "=" * 100)
print("Final Hybrid Evaluation Results")
print("=" * 100)

for task in FINAL_TASK_FIELDS:
    result = final_eval_results[task]

    print("\n" + "=" * 80)
    print(task)
    print("=" * 80)
    print("Weight:", result["weight"])
    print("Macro F1:", round(result["macro_f1"], 4))
    print("Micro F1:", round(result["micro_f1"], 4))
    print("Accuracy:", round(result["accuracy"], 4))
    print("Weighted Score:", round(result["weighted_score"], 6))

    y_true = [
        item[f"{task}_true"]
        for item in hybrid_final_predictions
    ]

    y_pred = [
        item[f"{task}_pred"]
        for item in hybrid_final_predictions
    ]

    print("\nClassification Report:")
    print(
        classification_report(
            y_true,
            y_pred,
            labels=LABEL_LISTS[task],
            zero_division=0
        )
    )

print("\n" + "=" * 100)
print("Final Official Weighted Score")
print("=" * 100)
print("Final Total Score:", round(final_total_score, 6))

final_summary_rows = []

for task in FINAL_TASK_FIELDS:
    result = final_eval_results[task]

    final_summary_rows.append({
        "task": task,
        "method": (
            "RoBERTa ensemble"
            if task in ["promise_status", "evidence_status"]
            else f"GPT + RAG ({GPT_MODEL})"
        ),
        "macro_f1": result["macro_f1"],
        "micro_f1": result["micro_f1"],
        "accuracy": result["accuracy"],
        "official_weight": result["weight"],
        "weighted_score": result["weighted_score"]
    })

final_summary_df = pd.DataFrame(final_summary_rows)

print("\nFinal Summary Table:")
display(final_summary_df)

FINAL_HYBRID_PREDICTIONS_PATH = os.path.join(
    OUTPUT_DIR,
    "final_hybrid_predictions.json"
)

FINAL_HYBRID_EVAL_PATH = os.path.join(
    OUTPUT_DIR,
    "final_hybrid_eval_results.json"
)

FINAL_HYBRID_SUMMARY_CSV_PATH = os.path.join(
    OUTPUT_DIR,
    "final_hybrid_eval_summary.csv"
)

FINAL_HYBRID_PREDICTIONS_CSV_PATH = os.path.join(
    OUTPUT_DIR,
    "final_hybrid_predictions.csv"
)

final_eval_output = {
    "final_total_score": final_total_score,
    "task_results": final_eval_results,
    "summary": final_summary_rows,
    "num_predictions": len(hybrid_final_predictions),
    "roberta_model_dir": ROBERTA_MODEL_DIR,
    "gpt_model": GPT_MODEL,
    "top_k": BEST_TOP_K
}

with open(FINAL_HYBRID_PREDICTIONS_PATH, "w", encoding="utf-8") as f:
    json.dump(hybrid_final_predictions, f, ensure_ascii=False, indent=2)

with open(FINAL_HYBRID_EVAL_PATH, "w", encoding="utf-8") as f:
    json.dump(final_eval_output, f, ensure_ascii=False, indent=2)

final_summary_df.to_csv(
    FINAL_HYBRID_SUMMARY_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

pd.DataFrame(hybrid_final_predictions).to_csv(
    FINAL_HYBRID_PREDICTIONS_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

print("\n✅ 最終四任務預測已儲存：")
print(FINAL_HYBRID_PREDICTIONS_PATH)

print("\n✅ 最終四任務評估結果已儲存：")
print(FINAL_HYBRID_EVAL_PATH)

print("\n✅ 最終 summary CSV 已儲存：")
print(FINAL_HYBRID_SUMMARY_CSV_PATH)

print("\n✅ 最終逐筆預測 CSV 已儲存：")
print(FINAL_HYBRID_PREDICTIONS_CSV_PATH)

print("\n✅ Cell 27 完成：最終 Hybrid 官方總分計算完成")
````

### 任務依賴規則後處理

**執行類型：必跑；套用四任務前後關係。**

這格會強制套用任務間的標籤依賴：`promise_status = No` 時後三個任務都為 `N/A`；`evidence_status = No` 時 `evidence_quality = N/A`。


````python
FINAL_HYBRID_PREDICTIONS_PATH = os.path.join(
    OUTPUT_DIR,
    "final_hybrid_predictions.json"
)

POSTPROCESS_PREDICTIONS_PATH = os.path.join(
    OUTPUT_DIR,
    "final_hybrid_predictions_postprocessed.json"
)

POSTPROCESS_EVAL_PATH = os.path.join(
    OUTPUT_DIR,
    "final_hybrid_eval_results_postprocessed.json"
)

print("讀取 hybrid 預測：")
print(FINAL_HYBRID_PREDICTIONS_PATH)

if not os.path.exists(FINAL_HYBRID_PREDICTIONS_PATH):
    raise FileNotFoundError(
        f"❌ 找不到 final_hybrid_predictions.json：{FINAL_HYBRID_PREDICTIONS_PATH}"
    )

with open(FINAL_HYBRID_PREDICTIONS_PATH, "r", encoding="utf-8") as f:
    hybrid_final_predictions = json.load(f)

original_score, original_eval_results = compute_final_score_from_predictions(
    hybrid_final_predictions,
    title="Original Hybrid Evaluation"
)

postprocessed_predictions = [
    apply_task_dependency_rules(item)
    for item in hybrid_final_predictions
]

changed_records = [
    item for item in postprocessed_predictions
    if item.get("dependency_rule_applied")
]

print("\n任務依賴規則修正筆數:", len(changed_records))

if len(changed_records) > 0:
    change_rows = []
    for item in changed_records:
        for change in item["dependency_rule_changes"]:
            change_rows.append({
                "id": item["id"],
                "field": change["field"],
                "before": change["before"],
                "after": change["after"],
                "reason": change["reason"],
            })

    dependency_change_df = pd.DataFrame(change_rows)
    display(dependency_change_df.head(30))

postprocessed_score, post_eval_results = compute_final_score_from_predictions(
    postprocessed_predictions,
    title="Post-processed Hybrid Evaluation"
)

post_eval_output = {
    "rule_name": "task_dependency_rules",
    "rules": [
        "if promise_status_pred == No: verification_timeline_pred = N/A, evidence_status_pred = N/A, evidence_quality_pred = N/A",
        "if evidence_status_pred == No: evidence_quality_pred = N/A",
    ],
    "original_score": original_score,
    "postprocessed_score": postprocessed_score,
    "score_delta": postprocessed_score - original_score,
    "original_eval_results": original_eval_results,
    "post_eval_results": post_eval_results,
    "changed_record_count": len(changed_records),
}

with open(POSTPROCESS_PREDICTIONS_PATH, "w", encoding="utf-8") as f:
    json.dump(postprocessed_predictions, f, ensure_ascii=False, indent=2)

with open(POSTPROCESS_EVAL_PATH, "w", encoding="utf-8") as f:
    json.dump(post_eval_output, f, ensure_ascii=False, indent=2)

print("\n✅ 後處理預測已儲存：")
print(POSTPROCESS_PREDICTIONS_PATH)
print("\n✅ 後處理評估結果已儲存：")
print(POSTPROCESS_EVAL_PATH)
print("Score Delta:", round(post_eval_output["score_delta"], 6))
````

這一段用來檢查模型行為、錯誤樣本與 prompt 內容。它有助於分析結果，但不是每次正式推論都一定要完整執行。


### 錯誤分析

**執行類型：建議跑；產出錯誤分析。**

整理四個任務的錯誤樣本、混淆矩陣與 Misleading 檢查，作為後續調整依據。


````python
FINAL_HYBRID_PREDICTIONS_PATH = os.path.join(
    OUTPUT_DIR,
    "final_hybrid_predictions.json"
)

print("讀取最終四任務預測：")
print(FINAL_HYBRID_PREDICTIONS_PATH)

if not os.path.exists(FINAL_HYBRID_PREDICTIONS_PATH):
    raise FileNotFoundError(
        f"❌ 找不到 final_hybrid_predictions.json：{FINAL_HYBRID_PREDICTIONS_PATH}"
    )

with open(FINAL_HYBRID_PREDICTIONS_PATH, "r", encoding="utf-8") as f:
    hybrid_final_predictions = json.load(f)

print("final predictions 筆數:", len(hybrid_final_predictions))

if len(hybrid_final_predictions) != len(test_data):
    print("⚠️ final predictions 筆數與 test_data 不一致")
else:
    print("✅ final predictions 筆數正確")

final_df = pd.DataFrame(hybrid_final_predictions)

print("\nfinal_df shape:", final_df.shape)
display(final_df.head())

FINAL_TASK_FIELDS = [
    "promise_status",
    "verification_timeline",
    "evidence_status",
    "evidence_quality"
]

ERROR_ANALYSIS_DIR = os.path.join(
    OUTPUT_DIR,
    "error_analysis"
)

os.makedirs(ERROR_ANALYSIS_DIR, exist_ok=True)

print("\nERROR_ANALYSIS_DIR:", ERROR_ANALYSIS_DIR)

error_summary_rows = {}

for task in FINAL_TASK_FIELDS:
    true_col = f"{task}_true"
    pred_col = f"{task}_pred"

    task_df = final_df.copy()

    task_df[f"{task}_is_correct"] = (
        task_df[true_col] == task_df[pred_col]
    )

    error_df = task_df[
        task_df[f"{task}_is_correct"] == False
    ].copy()

    correct_count = int(task_df[f"{task}_is_correct"].sum())
    error_count = int(len(error_df))
    total_count = int(len(task_df))

    print("\n" + "=" * 100)
    print(f"Task: {task}")
    print("=" * 100)
    print("Total:", total_count)
    print("Correct:", correct_count)
    print("Error:", error_count)
    print("Accuracy:", round(correct_count / total_count, 4))

    error_df["error_type"] = (
        error_df[true_col].astype(str)
        + " -> "
        + error_df[pred_col].astype(str)
    )

    error_type_counter = Counter(error_df["error_type"])

    print("\n錯誤類型統計：")
    for error_type, count in error_type_counter.most_common():
        print(f"{error_type}: {count}")

    output_cols = [
        "id",
        "data",
        true_col,
        pred_col,
        "error_type"
    ]

    for other_task in FINAL_TASK_FIELDS:
        if other_task != task:
            output_cols.extend([
                f"{other_task}_true",
                f"{other_task}_pred"
            ])

    output_cols = [
        col for col in output_cols
        if col in error_df.columns
    ]

    error_output_path = os.path.join(
        ERROR_ANALYSIS_DIR,
        f"{task}_errors.csv"
    )

    error_df[output_cols].to_csv(
        error_output_path,
        index=False,
        encoding="utf-8-sig"
    )

    print("\n✅ 錯誤分析 CSV 已儲存：")
    print(error_output_path)

    error_summary_rows[task] = {
        "task": task,
        "total_count": total_count,
        "correct_count": correct_count,
        "error_count": error_count,
        "accuracy": correct_count / total_count,
        "top_error_types": dict(error_type_counter.most_common(10)),
        "error_csv_path": error_output_path
    }

confusion_outputs = {}

for task in FINAL_TASK_FIELDS:
    true_col = f"{task}_true"
    pred_col = f"{task}_pred"
    labels = LABEL_LISTS[task]

    y_true = final_df[true_col].tolist()
    y_pred = final_df[pred_col].tolist()

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=labels
    )

    cm_df = pd.DataFrame(
        cm,
        index=[f"true_{label}" for label in labels],
        columns=[f"pred_{label}" for label in labels]
    )

    print("\n" + "=" * 100)
    print(f"Confusion Matrix: {task}")
    print("=" * 100)
    display(cm_df)

    cm_output_path = os.path.join(
        ERROR_ANALYSIS_DIR,
        f"{task}_confusion_matrix.csv"
    )

    cm_df.to_csv(
        cm_output_path,
        encoding="utf-8-sig"
    )

    confusion_outputs[task] = {
        "labels": labels,
        "matrix": cm.tolist(),
        "csv_path": cm_output_path
    }

    print("✅ Confusion matrix 已儲存：")
    print(cm_output_path)

print("\n" + "=" * 100)
print("Misleading 特別檢查")
print("=" * 100)

misleading_true_df = final_df[
    final_df["evidence_quality_true"] == "Misleading"
].copy()

misleading_pred_df = final_df[
    final_df["evidence_quality_pred"] == "Misleading"
].copy()

print("True Misleading 筆數:", len(misleading_true_df))
print("Pred Misleading 筆數:", len(misleading_pred_df))

print("\nTrue Misleading 資料：")
display(
    misleading_true_df[
        [
            "id",
            "data",
            "evidence_quality_true",
            "evidence_quality_pred",
            "promise_status_true",
            "promise_status_pred",
            "evidence_status_true",
            "evidence_status_pred",
            "verification_timeline_true",
            "verification_timeline_pred"
        ]
    ]
)

print("\nPred Misleading 資料：")
display(
    misleading_pred_df[
        [
            "id",
            "data",
            "evidence_quality_true",
            "evidence_quality_pred",
            "promise_status_true",
            "promise_status_pred",
            "evidence_status_true",
            "evidence_status_pred",
            "verification_timeline_true",
            "verification_timeline_pred"
        ]
    ]
)

MISLEADING_TRUE_PATH = os.path.join(
    ERROR_ANALYSIS_DIR,
    "misleading_true_records.csv"
)

MISLEADING_PRED_PATH = os.path.join(
    ERROR_ANALYSIS_DIR,
    "misleading_pred_records.csv"
)

misleading_true_df.to_csv(
    MISLEADING_TRUE_PATH,
    index=False,
    encoding="utf-8-sig"
)

misleading_pred_df.to_csv(
    MISLEADING_PRED_PATH,
    index=False,
    encoding="utf-8-sig"
)

print("\n✅ True Misleading CSV 已儲存：")
print(MISLEADING_TRUE_PATH)

print("\n✅ Pred Misleading CSV 已儲存：")
print(MISLEADING_PRED_PATH)

timeline_errors = final_df[
    final_df["verification_timeline_true"]
    != final_df["verification_timeline_pred"]
].copy()

quality_errors = final_df[
    final_df["evidence_quality_true"]
    != final_df["evidence_quality_pred"]
].copy()

print("\n" + "=" * 100)
print("GPT + RAG 錯誤重點")
print("=" * 100)

print("verification_timeline 錯誤筆數:", len(timeline_errors))
print("evidence_quality 錯誤筆數:", len(quality_errors))

print("\nverification_timeline 前 10 筆錯誤：")
display(
    timeline_errors[
        [
            "id",
            "data",
            "verification_timeline_true",
            "verification_timeline_pred",
            "promise_status_pred",
            "evidence_status_pred",
            "evidence_quality_true",
            "evidence_quality_pred"
        ]
    ].head(10)
)

print("\nevidence_quality 前 10 筆錯誤：")
display(
    quality_errors[
        [
            "id",
            "data",
            "evidence_quality_true",
            "evidence_quality_pred",
            "promise_status_pred",
            "evidence_status_pred",
            "verification_timeline_true",
            "verification_timeline_pred"
        ]
    ].head(10)
)

ERROR_ANALYSIS_SUMMARY_PATH = os.path.join(
    ERROR_ANALYSIS_DIR,
    "error_analysis_summary.json"
)

error_analysis_summary = {
    "final_total_score": final_total_score if "final_total_score" in globals() else None,
    "task_error_summary": error_summary_rows,
    "confusion_matrices": confusion_outputs,
    "misleading": {
        "true_misleading_count": len(misleading_true_df),
        "pred_misleading_count": len(misleading_pred_df),
        "true_misleading_csv": MISLEADING_TRUE_PATH,
        "pred_misleading_csv": MISLEADING_PRED_PATH
    }
}

with open(ERROR_ANALYSIS_SUMMARY_PATH, "w", encoding="utf-8") as f:
    json.dump(error_analysis_summary, f, ensure_ascii=False, indent=2)

print("\n✅ 錯誤分析 summary 已儲存：")
print(ERROR_ANALYSIS_SUMMARY_PATH)

print("\n✅ Cell 28 完成：四任務錯誤分析與 Misleading 檢查完成")
````

````python
new_test_dataset = ESGRoBERTaDataset(
    data=test_for_model,
    tokenizer=tokenizer,
    max_len=MAX_LEN,
    task_fields=ROBERTA_TASK_FIELDS,
    label2id=label2id,
    text_field="input_text"
)

new_test_loader = DataLoader(
    new_test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

batch = next(iter(new_test_loader))

print("new_test_dataset:", len(new_test_dataset))
print("new_test_loader batches:", len(new_test_loader))
print("input_ids shape:", batch["input_ids"].shape)
print("has labels:", "labels" in batch)
````

### 無標籤資料 RoBERTa Ensemble 推論

**執行類型：必跑；產生 Task 1 / Task 3 預測。**

載入已訓練好的三個 seed checkpoint，對新 test set 做 promise_status 與 evidence_status 預測；結果只保留在記憶體中供最後合併使用。


````python
ENSEMBLE_SEEDS = [42, 123, 2024]
ENSEMBLE_MODEL_PATHS = [
    os.path.join(
        ROBERTA_MODEL_DIR,
        f"best_roberta_seed_{seed}.pt"
    )
    for seed in ENSEMBLE_SEEDS
]

for seed, path in zip(ENSEMBLE_SEEDS, ENSEMBLE_MODEL_PATHS):
    print(f"Seed {seed}: {path}")
    if not os.path.exists(path):
        raise FileNotFoundError(f"❌ 找不到模型檔案：{path}")

new_roberta_ensemble_result = predict_roberta_ensemble_memory_safe(
    model_paths=ENSEMBLE_MODEL_PATHS,
    data_loader=new_test_loader,
    device=device
)

new_roberta_predictions = []

for i, item_id in enumerate(new_roberta_ensemble_result["all_ids"]):
    record = {
        "id": normalize_id(item_id),
        "text": new_roberta_ensemble_result["all_texts"][i],
    }

    for task in ROBERTA_TASK_FIELDS:
        pred_id = new_roberta_ensemble_result["all_pred_labels"][task][i]
        pred_label = id2label[task][pred_id]
        probs = new_roberta_ensemble_result["all_pred_probs"][task][i]

        record[f"{task}_pred"] = pred_label
        record[f"{task}_pred_id"] = int(pred_id)
        record[f"{task}_probs"] = {
            id2label[task][label_id]: float(prob)
            for label_id, prob in enumerate(probs)
        }

    new_roberta_predictions.append(record)

roberta_pred_by_id = {
    normalize_id(item["id"]): {
        "id": normalize_id(item["id"]),
        "promise_status": item["promise_status_pred"],
        "evidence_status": item["evidence_status_pred"],
        "promise_status_probs": item.get("promise_status_probs"),
        "evidence_status_probs": item.get("evidence_status_probs"),
    }
    for item in new_roberta_predictions
}

print("✅ 新測試集 RoBERTa 預測完成")
print("預測筆數:", len(new_roberta_predictions))
display(pd.DataFrame(new_roberta_predictions).head())
````

### 無標籤資料 GPT/RAG v4 推論

**執行類型：必跑；產生 Task 2 / Task 4 預測。**

使用 retrieval examples、RoBERTa 輔助預測與 v4 prompt，對新 test set 預測 verification_timeline 與 evidence_quality；正式預測檔只在最後合併輸出。


````python
NEW_GPT_RAG_PROGRESS_PATH = os.path.join(
    OUTPUT_DIR,
    "new_test_gpt_rag_task24_progress.json"
)

RESET_NEW_GPT_PROGRESS = False

if RESET_NEW_GPT_PROGRESS and os.path.exists(NEW_GPT_RAG_PROGRESS_PATH):
    os.remove(NEW_GPT_RAG_PROGRESS_PATH)
    print("✅ 已刪除新測試集 GPT 進度檔")

if os.path.exists(NEW_GPT_RAG_PROGRESS_PATH):
    with open(NEW_GPT_RAG_PROGRESS_PATH, "r", encoding="utf-8") as f:
        new_gpt_rag_predictions = json.load(f)
    print("✅ 從進度檔續跑:", NEW_GPT_RAG_PROGRESS_PATH)
else:
    new_gpt_rag_predictions = []

done_ids = {
    normalize_id(item["id"])
    for item in new_gpt_rag_predictions
}

SAVE_EVERY = 1
SLEEP_BETWEEN_ITEMS = 0.2

for idx, item in enumerate(tqdm(test_data, desc="New test GPT + RAG")):
    item_id = normalize_id(item["id"])

    if item_id in done_ids:
        continue

    print("\n" + "=" * 100)
    print(f"正在預測新測試資料第 {idx + 1}/{len(test_data)} 筆 | id = {item_id}")
    print("=" * 100)

    try:
        timeline_result = predict_timeline_with_gpt_rag(
            test_item=item,
            top_k=BEST_TOP_K
        )

        quality_result = predict_quality_with_gpt_rag(
            test_item=item,
            top_k=BEST_TOP_K
        )

        roberta_pred = roberta_pred_by_id[item_id]

        pred_record = {
            "id": item_id,
            "data": item["data"],
            "verification_timeline_pred": timeline_result["prediction"],
            "evidence_quality_pred": quality_result["prediction"],
            "promise_status_roberta_pred": roberta_pred["promise_status"],
            "evidence_status_roberta_pred": roberta_pred["evidence_status"],
            "top_k": BEST_TOP_K,
            "timeline_retrieved_ids": timeline_result["retrieved_ids"],
            "timeline_retrieved_similarities": timeline_result["retrieved_similarities"],
            "quality_retrieved_ids": quality_result["retrieved_ids"],
            "quality_retrieved_similarities": quality_result["retrieved_similarities"],
            "timeline_raw_response": timeline_result["raw_response"],
            "quality_raw_response": quality_result["raw_response"],
            "timeline_parsed_response": timeline_result["parsed_response"],
            "quality_parsed_response": quality_result["parsed_response"],
            "timeline_prompt_length": timeline_result["prompt_length"],
            "quality_prompt_length": quality_result["prompt_length"],
            "gpt_model": GPT_MODEL
        }

        new_gpt_rag_predictions.append(pred_record)
        done_ids.add(item_id)

        print("✅ 預測完成")
        print("Timeline pred:", pred_record["verification_timeline_pred"])
        print("Quality pred:", pred_record["evidence_quality_pred"])

        if len(new_gpt_rag_predictions) % SAVE_EVERY == 0:
            with open(NEW_GPT_RAG_PROGRESS_PATH, "w", encoding="utf-8") as f:
                json.dump(new_gpt_rag_predictions, f, ensure_ascii=False, indent=2)

        time.sleep(SLEEP_BETWEEN_ITEMS)

    except Exception as e:
        print("❌ 此筆預測失敗")
        print("id:", item_id)
        print("錯誤訊息:", e)

        error_record = {
            "id": item_id,
            "data": item["data"],
            "verification_timeline_pred": "N/A",
            "evidence_quality_pred": "N/A",
            "promise_status_roberta_pred": roberta_pred_by_id.get(item_id, {}).get("promise_status"),
            "evidence_status_roberta_pred": roberta_pred_by_id.get(item_id, {}).get("evidence_status"),
            "top_k": BEST_TOP_K,
            "gpt_model": GPT_MODEL,
            "error": str(e)
        }

        new_gpt_rag_predictions.append(error_record)
        done_ids.add(item_id)

        with open(NEW_GPT_RAG_PROGRESS_PATH, "w", encoding="utf-8") as f:
            json.dump(new_gpt_rag_predictions, f, ensure_ascii=False, indent=2)

print("\n✅ 新測試集 GPT/RAG 預測完成")
print("預測筆數:", len(new_gpt_rag_predictions))
````

### 無標籤資料最終預測輸出

**執行類型：必跑；產生最終四任務預測。**

整合 RoBERTa Task 1/3 與 GPT/RAG Task 2/4，套用任務依賴規則後，只輸出最終合併 JSON 與 CSV。


````python
NEW_FINAL_PRED_PATH = os.path.join(
    OUTPUT_DIR,
    "new_test_final_predictions.json"
)

NEW_FINAL_PRED_CSV_PATH = os.path.join(
    OUTPUT_DIR,
    "new_test_final_predictions.csv"
)

roberta_by_id = {
    normalize_id(item["id"]): item
    for item in new_roberta_predictions
}

gpt_by_id = {
    normalize_id(item["id"]): item
    for item in new_gpt_rag_predictions
}

missing_roberta_ids = set(test_by_id.keys()) - set(roberta_by_id.keys())
missing_gpt_ids = set(test_by_id.keys()) - set(gpt_by_id.keys())

if missing_roberta_ids:
    raise ValueError(f"❌ 有新測試資料找不到 RoBERTa 預測：{list(missing_roberta_ids)[:10]}")

if missing_gpt_ids:
    raise ValueError(f"❌ 有新測試資料找不到 GPT/RAG 預測：{list(missing_gpt_ids)[:10]}")

new_final_predictions = []

for item in test_data:
    item_id = normalize_id(item["id"])
    roberta_item = roberta_by_id[item_id]
    gpt_item = gpt_by_id[item_id]

    final_item = {
        "id": item_id,
        "data": item["data"],
        "promise_status_pred": roberta_item["promise_status_pred"],
        "verification_timeline_pred": gpt_item["verification_timeline_pred"],
        "evidence_status_pred": roberta_item["evidence_status_pred"],
        "evidence_quality_pred": gpt_item["evidence_quality_pred"],
        "promise_status_method": "RoBERTa ensemble",
        "verification_timeline_method": f"GPT + RAG v4 ({gpt_item.get('gpt_model', GPT_MODEL)})",
        "evidence_status_method": "RoBERTa ensemble",
        "evidence_quality_method": f"GPT + RAG v4 ({gpt_item.get('gpt_model', GPT_MODEL)})",
        "top_k": gpt_item.get("top_k", BEST_TOP_K),
        "gpt_model": gpt_item.get("gpt_model", GPT_MODEL),
    }

    final_item = apply_task_dependency_rules(final_item)
    new_final_predictions.append(final_item)

with open(NEW_FINAL_PRED_PATH, "w", encoding="utf-8") as f:
    json.dump(new_final_predictions, f, ensure_ascii=False, indent=2)

new_final_df = pd.DataFrame(new_final_predictions)
new_final_df.to_csv(
    NEW_FINAL_PRED_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

print("✅ 新測試集最終預測完成")
print("JSON:", NEW_FINAL_PRED_PATH)
print("CSV:", NEW_FINAL_PRED_CSV_PATH)
print("預測筆數:", len(new_final_predictions))

changed_count = int(new_final_df["dependency_rule_applied"].sum())
print("任務依賴規則修正筆數:", changed_count)

display(new_final_df.head())
````
